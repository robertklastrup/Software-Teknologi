## Question 5 — Eigendecomposition (Diagonalization)

Given matrix:

$$B = \begin{bmatrix} 6 & 2 & 0 \\ 2 & 3 & 0 \\ 0 & 0 & 5 \end{bmatrix}$$

1. **Diagonalize** matrix **B**. Print **V**, **D**, and **V⁻¹**.
2. Reconstruct **B** from **V**, **D**, **V⁻¹** and verify the result.


````python
# Question 5
B = np.array([[6, 2, 0],
              [2, 3, 0],
              [0, 0, 5]], dtype=float)

# 1. Diagonalize
# your code
eigenvalues, V = np.linalg.eig(B)
D   = np.diag(eigenvalues)
V_inv = np.linalg.inv(V)

print ("V: \n", np.round(V,4), "\nD: \n", np.round(D,4), "\nV^1: \n", np.round(V_inv,4), '\n')

# 2. Reconstruct and verify
# your code
B_rec = V @ D @ V_inv
print ("Reconstruction: \n", B_rec)
print ("The matrix B: \n", B)
````

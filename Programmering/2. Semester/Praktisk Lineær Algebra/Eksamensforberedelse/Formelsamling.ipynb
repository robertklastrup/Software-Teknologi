## Vector

### Norm af en vector
- Normen af et array (eller en vektor) er et matematisk mål for arrayets længde, størrelse eller "magnitude". Kort sagt fortæller normen dig, hvor "kraftfuldt" et array er, eller hvad den samlede afstand er fra et nulpunkt til punktet defineret af arrayets værdier

Opgave eksempel:
Givne vektorer **v** og **w** er defineret som:

$$v = \begin{bmatrix} 4 \\ 3 \end{bmatrix}, \quad w = \begin{bmatrix} 5 \\ 2 \end{bmatrix}$$

Beregn normen af vector v:

```python
v = ([4,3])
norm_v = np.linalg.norm(v)
print ("The norm of v = ", norm_v)
```

### Unit vector
- Enhedsvektorer bruges til at beskrive, hvilken vej en given retning peger, uden at tage højde for længde.
- Man kan forestille sig en pil, hvor retningen bestemmer vinklen, og enhedsvektoren blot er selve spidsen af pilen

Nu bruger vi vector v fra tidligere opgave:

```python
unit_vector_v=(v/norm_v)
print("The unit vector is: ",'\n', unit_vector_v)
```

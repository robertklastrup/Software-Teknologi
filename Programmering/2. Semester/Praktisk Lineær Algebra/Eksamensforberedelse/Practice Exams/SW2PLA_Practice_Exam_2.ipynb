## Question 5 — Eigendecomposition (Diagonalization)

Given matrix:

$$B = \begin{bmatrix} 6 & 2 & 0 \\ 2 & 3 & 0 \\ 0 & 0 & 5 \end{bmatrix}$$

1. **Diagonalize** matrix **B**. Print **V**, **D**, and **V⁻¹**.
2. Reconstruct **B** from **V**, **D**, **V⁻¹** and verify the result.


````python
# Question 5
B = np.array([[6, 2, 0],
              [2, 3, 0],
              [0, 0, 5]], dtype=float)

# 1. Diagonalize
# your code
eigenvalues, V = np.linalg.eig(B)
D   = np.diag(eigenvalues)
V_inv = np.linalg.inv(V)

print ("V: \n", np.round(V,4), "\nD: \n", np.round(D,4), "\nV^1: \n", np.round(V_inv,4), '\n')

# 2. Reconstruct and verify
# your code
B_rec = V @ D @ V_inv
print ("Reconstruction: \n", B_rec)
print ("The matrix B: \n", B)
````

## Question 6 — Singular Value Decomposition (SVD)

Using matrix **B** from Question 5:

1. Compute the **SVD** of **B**. Print **U**, **S**, and **Vt**.
2. Reconstruct **B** from **U**, **S**, **Vt** and verify.
3. Compute and print a **rank-2 low-rank approximation** of **B** and print the approximation error (Frobenius norm).


````python
# Question 6
# 1. Compute SVD
# your code

U, S, Vt = np.linalg.svd(B)

print ("U: \n", np.round(U,4), "\nS: \n", np.round (S,4), "\n Vt: \n", np.round(Vt,4), '\n')


# 2. Reconstruct and verify
# your code
Sigma = np.zeros_like(B)
np.fill_diagonal(Sigma, S)
B_rec2 = U @ Sigma @ Vt
print("\nVerifikation U @ Sigma @ Vt ≈ B:"); print(np.round(B_rec2, 6))
print ("B: \n", B)

# 3. Rank-2 approximation + error
# your code
k = 2
B_approx = sum(S[i] * np.outer(U[:,i], Vt[i,:]) for i in range(k))
print(f"\nLow-rank approximation (k={k}):"); print(np.round(B_approx, 4))
print(f"Error (Frobenius): {np.linalg.norm(B - B_approx):.4f}", "\n")

print("The Frobenius norm is a size of a matrix. \n",
"It is used to measure the amount of information that is lost (in this case) \n",
"a rank-2 approximation. The lower the better",'\n'
"< 1% is very good \n1-5% is acceptable \n > 10% big information loss")
print ("Here it is in %:")
error_rel = np.linalg.norm(B - B_approx) / np.linalg.norm(B) * 100
print("Approksimationsfejl: ", error_rel, "%")

````

## Question 7 — Generalized Linear Model (GLM / Least Squares)

A researcher measures outdoor temperature and ice cream sales:

| Temperature (°C) | Ice cream sales |
|:---:|:---:|
| 14 | 215 |
| 18 | 325 |
| 20 | 370 |
| 24 | 430 |
| 26 | 500 |
| 16 | 280 |
| 22 | 405 |
| 28 | 545 |

1. Use **least squares** to fit a linear model. Print the equation `y = β1*x + β0`.
2. Print the **predicted sales** when temperature is **25°C**.
3. Print the **SSE** and **R²**.


````python
# Question 7
temp  = np.array([14, 18, 20, 24, 26, 16, 22, 28], dtype=float)
sales = np.array([215, 325, 370, 430, 500, 280, 405, 545], dtype=float)

# 1. Fit linear model
# your code
X = np.column_stack([temp, np.ones(len(temp))])
beta, _, _, _ = np.linalg.lstsq(X, sales, rcond=None)
beta1, beta0 = beta
print(f"Model: y = {beta1:.4f}x + {beta0:.4f}")


# 2. Predicted sales at 25°C
# your code
x_new = 25
y_pred = beta1 * x_new + beta0
print(f"Forudsagt score for {x_new} timer: {y_pred:.2f}")

# 3. SSE and R²
# your code
y_hat = X @ beta
SSE = np.sum((sales - y_hat)**2)
SS_tot = np.sum((sales - np.mean(sales))**2)
R2 = 1 - SSE / SS_tot
print(f"SSE: {SSE:.4f}")
print(f"R²:  {R2:.4f}  (1.0 = perfekt fit)")
````

4. In your own words: what does it mean that the nullspace has dimension > 0?


### Answer to Question 8.4

*(Write your answer here in your own words)*


````python
# Question 8
M = np.array([[1,  2,  4,  8],
              [2,  4,  8, 16],
              [3,  6, 12, 24]], dtype=float)

# 1. Rank
# your code
rank = np.linalg.matrix_rank(M)
print ("Rank: ", rank)


# 2. Dim(nullspace)
# your code
nullity = M.shape[1] - rank
print ("Dimension of nullspace: ", nullity)

# 3. Does v = [2, -1, 0, 0] belong to nullspace?
v = np.array([2, -1, 0, 0], dtype=float)
# your code
Mv = M @ v
print("\nAv =", np.round(Mv, 10))
print("v tilhører nullspace:", np.allclose(Mv, 0))
````

## Question 9 — Basis Transformation *(GAI er ikke tilladt)*

In $\mathbb{R}^3$, consider two bases:

$$B = \begin{bmatrix} 1 & 0 & 1 \\ 0 & 1 & 1 \\ 1 & 1 & 0 \end{bmatrix}, \quad
F = \begin{bmatrix} 2 & 1 & 0 \\ 1 & 2 & 1 \\ 0 & 1 & 2 \end{bmatrix}$$

A vector $v$ has coordinates $(3, 1, 2)$ in basis **B**.

1. Find the coordinates of $v$ in the **standard basis**.
2. Find the coordinates of $v$ in basis **F**.
3. Verify your answer.

In your own words: what does it mean to change basis?


### Answer to Question 9.4

*(Write your answer here in your own words)*


````python
# Question 9
B = np.array([[1, 0, 1],
              [0, 1, 1],
              [1, 1, 0]], dtype=float)

F = np.array([[2, 1, 0],
              [1, 2, 1],
              [0, 1, 2]], dtype=float)

c_B = np.array([3, 1, 2], dtype=float)

# 1. Coordinates in standard basis
# your code
v_std = B @ c_B
print("Koordinater i standardbasis:", v_std)


# 2. Coordinates in basis F
# your code
c_F = np.linalg.solve(F, v_std)
print("Koordinater i basis F:", np.round(c_F, 6))

# 3. Verification
# your code
print("Verifikation F @ c_F:", np.round(F @ c_F, 6))

````

At skifte basis betyder at man beskriver den samme vektor med et nyt koordinatsystem. Vektoren selv ændrer sig ikke — kun de tal vi bruger til at beskrive den. Det svarer til at måle den samme afstand i km i stedet for miles: det er stadig den samme afstand, men tallene er anderledes.

## Question 11 — PCA *(GAI er ikke tilladt)*

Below is a small dataset with 3 features:

```python
X = np.array([[2.5, 2.4, 3.1],
              [0.5, 0.7, 1.2],
              [2.2, 2.9, 2.8],
              [1.9, 2.2, 2.5],
              [3.1, 3.0, 3.6],
              [2.3, 2.7, 2.9],
              [2.0, 1.6, 2.3],
              [1.0, 1.1, 1.5],
              [1.5, 1.6, 1.8],
              [1.1, 0.9, 1.3]], dtype=float)
```

Implement PCA using the **5 steps**:
1. Mean-center the data
2. Compute the covariance matrix
3. Compute eigendecomposition
4. Sort by descending eigenvalues
5. Project data onto the first **2** principal components

Then print:
- The **explained variance** per component (%)
- The **shape** of the projected data

In your own words: what do the principal components represent?


### Answer — what do principal components represent?

*(Write your answer here in your own words)*


````python
# Question 11
X = np.array([[2.5, 2.4, 3.1],
              [0.5, 0.7, 1.2],
              [2.2, 2.9, 2.8],
              [1.9, 2.2, 2.5],
              [3.1, 3.0, 3.6],
              [2.3, 2.7, 2.9],
              [2.0, 1.6, 2.3],
              [1.0, 1.1, 1.5],
              [1.5, 1.6, 1.8],
              [1.1, 0.9, 1.3]], dtype=float)

# Step 1: Mean-center
# your code
X_centered = X - np.mean(X, axis=0, keepdims=True)

# Step 2: Covariance matrix
# your code
C = X_centered.T @ X_centered / (X_centered.shape[0] - 1)



# Step 3: Eigendecomposition
# your code
evals, evecs = np.linalg.eigh(C)



# Step 4: Sort descending
# your code
idx   = np.argsort(evals)[::-1]
evals = evals[idx]
evecs = evecs[:, idx]

# Step 5: Project onto 2 components
# your code
components = X @ evecs[:, :2]

# Print explained variance and shape
# your code
var_exp = evals / np.sum(evals) * 100
print ("Explained varians:")
for i, v in enumerate(var_exp):
    print(f"PC{i+1}: {v:.1f}%")

print("Shape of project onto 2 components:", components.shape)
````

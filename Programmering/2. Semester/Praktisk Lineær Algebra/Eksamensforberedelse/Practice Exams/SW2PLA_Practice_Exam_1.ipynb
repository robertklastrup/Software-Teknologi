## Question 5 — Eigendecomposition (Diagonalization)

Given matrix:

$$B = \begin{bmatrix} 5 & 2 & 0 \\ 2 & 3 & 0 \\ 0 & 0 & 4 \end{bmatrix}$$

1. **Diagonalize** matrix **B**. Print matrices **V**, **D**, and **V⁻¹**.
2. Use **V**, **D**, and **V⁻¹** to **reconstruct** matrix **B** and verify the result equals the original.


```python
# Question 5
B = np.array([[5, 2, 0],
              [2, 3, 0],
              [0, 0, 4]], dtype=float)

# 1. Diagonalize B → eigenvalues (D) and eigenvectors (V)
# your code
eigenvalues, V = np.linalg.eig(B)
D = np.diag(eigenvalues)
V_inv = np.linalg.inv(V)

print ("V: \n", V)
print ("D: \n", D)
print ("V^-1 is the inverse. \n", "The inverse of V is: \n",V_inv)



# 2. Reconstruct B = V @ D @ V_inv and print
# your code
B_verify = V @ D @ V_inv
print ("To verify that V, D and V^-1 is the same as B, we multiply them with each other",
       "The output when we multiply the matrixes is: \n", B_verify, '\n',
       "Here is the original matrix B: \n", B)



```

## Question 6 — Singular Value Decomposition (SVD)

## Question 7 — Generalized Linear Model (GLM / Least Squares)

The following dataset shows the number of hours a student studied and the resulting exam score:

| Hours studied | Exam score |
|:---:|:---:|
| 3 | 55 |
| 5 | 70 |
| 6 | 75 |
| 8 | 88 |
| 9 | 90 |
| 4 | 62 |
| 7 | 82 |

1. Use the **least squares** function to solve the GLM and print the resulting equation (e.g. `y = β1*x + β0`).
2. Print the **predicted score** for a student who studied **6.5 hours**.
3. Print the **SSE** (Sum of Squared Errors).


```python
# Question 7
hours  = np.array([3, 5, 6, 8, 9, 4, 7], dtype=float)
scores = np.array([55, 70, 75, 88, 90, 62, 82], dtype=float)

# 1. Build design matrix and solve using least squares
# your code
X = np.column_stack([hours, np.ones(len(hours))])  # Building the design matrix
result = np.linalg.lstsq(X, scores, rcond=None) #Solving using least squares
beta1, beta0 = result[0]
print(f"y = {beta1:.4f}*x + {beta0:.4f}")


# 2. Predicted score for 6.5 hours
# your code
predicted = beta1 * 6.5 + beta0
print(f"Predicted score for 6.5 hours: {predicted:.2f}")

# 3. SSE
# your code
scores_predicted = X @ result[0]
SSE = np.sum((scores - scores_predicted)**2)
print(f"SSE: {SSE:.4f}")
```

The mathematical expression that shows the reconstruction af A as a sum of all rank-1 layes of the SVD is:

It is good for composition, so that we have a lot fewer numbers to deal with.

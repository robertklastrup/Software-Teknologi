```python
#My take:
print ("My take: \n")
B = np.array ([[4, 0, 0],
               [1, 2, 1],
               [-3, 0, 3]])

eigenvalues, V = np.linalg.eig(B)
D   = np.diag(eigenvalues)
V_inv = np.linalg.inv(V)

print ("V: \n", np.round(V,4), "\nD: \n", np.round(D,4), "\nV^-1: \n", np.round(V_inv,4), '\n')

#To show the diagonalization is correct, we reconstruct and verify
B_rec = V @ D @ V_inv
print ("Reconstruction: \n", B_rec)
print ("The matrix B: \n", B, '\n')

print ("Solution sheet:")
B1 = np.array([[4,0,0],[1,2,1],[-3,0,3]])
evals, evecs = np.linalg.eig(B1)
V=evecs
print("V =", V)
D= np.diag(evals)
print("D =", D)
Vinv= np.linalg.inv(evecs)
print("Vinv =", Vinv, '\n')

# Option 1 Visualy inspecting it is the same result
print("Option 1: \n",V @ D @ Vinv)
# Option 2 Use isclose
print("Option 2: \n",np.isclose(B, V @ D @ Vinv ))
# Option 3 Use allclose
print("Option 3: \n",np.allclose(B, V @ D @ Vinv ))
```

```python
# 1. SVD:
U, S, Vt = np.linalg.svd(B)

print ("U: \n", np.round(U,4), "\nS: \n", np.round (S,4), "\n Vt: \n", np.round(Vt,4), '\n')

# 2. Reconstruct and verify
#First we convert S from a 1D array to a diagonal matrix:
S_matrix = np.diag(S)
#Now we reconstruct B:
B_verify2 = U @ S_matrix @ Vt
print ("B: \n", B , '\n')
print ("Verification (U @ S (matrix form) @ Vt) \n", np.round(B_verify2,2), '\n')

#Solution sheet
print ("Solution sheet:")

#1. SVD:
U,s,Vt = np.linalg.svd(B)
print(U)
S=np.diag(s)
print(S)
print(Vt)

#2. verify:
# Option 1 Visualy inspecting it is the same result
print(U @ S @ Vt)
# Option 2 Use isclose
print(np.isclose(B, U @ S @ Vt ))
# Option 3 Use allclose
print(np.allclose(B, U @ S @ Vt ))
```

2. Any matrix can be written as a sum of rank-1 matrices (outer products), weighted by the singular values. A low-rank approximation keeps only the $k$ most important terms, discarding the rest.

```python
#Q 4.1 in python:
k = 1
B_approx = sum(S[i] * np.outer(U[:, i], Vt[i, :]) for i in range(k))
print("Rank-1 approx:\n", np.round(B_approx, 4))
```

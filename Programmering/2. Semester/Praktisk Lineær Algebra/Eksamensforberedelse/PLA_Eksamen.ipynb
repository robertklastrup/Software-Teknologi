## Question 1

Given vectors v and w defined as:
v = [3, -6, 2] and w = [0, 4, -1]
1. Calculate and print the norm of vector v.
2. Create and print a unit vector associated with v.
3. Compute and print the projection of v onto w

```python
v = np.array([3, -6, 2])
w = np.array ([0, 4, -1])

#1 Norm of v
v_norm = np.linalg.norm(v)
print ("Norm af v: ", v_norm)

#2 unit vector of v
v_unit = v / v_norm
print("Enhedsvektor:", v_unit)
print("Verifikation — norm skal være 1:", np.round(np.linalg.norm(v_unit),2))

#3 projection of v onto w
#3 projection of v onto w
proj = (np.dot(v, w) / np.dot(w, w)) * w
print("Projektion af v onto w:", proj)

```

## Question 2

Given a matrix defined as:
A =
[[2, 1, 0, 3],
[0, 4, 1, 2],
[1, 0, 3, 1],
[3, 2, 1, 0],
[1, 1, 2, 4]]
1. Create and print the matrix A.
2. Compute and print the trace and the Frobenius norm of A^T A.
3. How many linearly independent columns does A have? Justify your answer based on the
shape of A and your result from 2.2

```python
#1 create and print matrix A
A = np.array([[2,1,0,3],
              [0,4,1,2],
              [1,0,3,1],
              [3,2,1,0],
              [1,1,2,4]])

print ("A: \n", A)

#2 Compute and print the trace and the Frobenius norm:
ATA = A.T @ A
print("A^T A:\n", ATA)
print("Trace of A^T A:", np.trace(ATA))
print("Frobenius norm of A^T A:", np.linalg.norm(ATA, 'fro'))

#3. How many linerly independent columns does A have:
rank = np.linalg.matrix_rank(A)
num_cols = A.shape[1]  # 4
print(f"Rank: {rank}, Number of columns: {num_cols}")
print(f"A has {rank} linearly independent columns.")
print(f"All columns independent: {rank == num_cols}")
```

Since the matrix A is a 5x4 matrix, it has for columns. Since rank = 4 is the same as n (number of columns), all columns are linearly independant.

## Question 3

1. Construct the left inverse R of matrix A defined in Question 2.
2. Confirm that the matrix R you computed is in fact a left inverse. If a left inverse does not
exist, explain why and construct the right inverse instead (and verify that it is indeed a
right inverse.)

```python
#1 left inverse R
R = np.linalg.inv(A.T @ A) @ A.T
print ("Left inverse R: \n", R)

#2 Confirm that the matrix R is the left inverse
print("R @ A:\n", np.round(R @ A, 6))
print("Is left inverse:", np.allclose(R @ A, np.eye(4)))

```

A left inverse exists because A is a 5×4 matrix with full column rank (rank = 4 = number of columns), so $(A^T A)$ is invertible and $R = (A^T A)^{-1} A^T$ is a valid left inverse.
A right inverse does not exist because A has more rows than columns (5 > 4), meaning A cannot span all of ℝ^5. A right inverse only exists when A has full row rank.

## Question 5

1. Diagonalize the matrix B:

B =

[[3, 1, 0],

[0, 3, 0],

[0, 1, 5]]

Print a matrix V, D, and V^-1. If the matrix is not diagonalizable, explain why.

2. Use V, D, and V^-1 to reconstruct B and verify that the reconstruction equals B

```python
B = np.array ([[3, 1, 0],
               [0, 3, 0],
               [0, 1, 5]])

eigenvalues, V = np.linalg.eig(B)
D   = np.diag(eigenvalues)
V_inv = np.linalg.inv(V)

print ("V: \n", np.round(V,4), "\nD: \n", np.round(D,4), "\nV^-1: \n", np.round(V_inv,4), '\n')

print ("Rank of V: ",np.linalg.matrix_rank(V))
print ("det of V: ",np.round(np.linalg.det(V),2))

# Check if V is invertible (required for diagonalization)
if np.linalg.matrix_rank(V) < B.shape[0]:
    print("B is NOT diagonalizable: eigenvector matrix V is singular.")
    print("Eigenvalue 3 has algebraic multiplicity 2 but geometric multiplicity 1.")
else:
    D = np.diag(eigenvalues)
    V_inv = np.linalg.inv(V)
    print("V:\n", np.round(V, 4))
    print("D:\n", np.round(D, 4))
    print("V_inv:\n", np.round(V_inv, 4))
    B_rec = V @ D @ V_inv
    print("Reconstruction matches B:", np.allclose(B_rec, B))


```

Since the rank of V (2) is not equal to n (3), and the determant is 0, V is singular which means B isn't diagonizable.

## Question 6

1. Compute the SVD of matrix B from Question 5. Print the resulting U, S, and V^T
matrices.
2. Use U, S, and V^T to reconstruct the matrix B and verify the reconstruction equals B

```python
#1. SVD
U, S, Vt = np.linalg.svd(B)

S_matrix = np.diag(S)
print ("U: \n", np.round(U,4), "\nS: \n", np.round (S_matrix,4), "\n Vt: \n", np.round(Vt,4), '\n')

# 2. Reconstruct and verify

B_verify = U @ S_matrix @ Vt
print ("B: \n", B , '\n')
print ("Verification (U @ S (matrix form) @ Vt) \n", np.round(B_verify,2), '\n')
print ("Equals B?: ", np.allclose(B,B_verify))
```

## Question 7

The table below contains an artificial dataset of student performance. We want to use a GLM to predict the final exam score given a student's weekly hours of self-study and their number of completed assignments.

| Hours of Self-Study / Week | Completed Assignments | Exam Score |
|:--------------------------:|:---------------------:|:----------:|
| 3  | 8  | 68 |
| 5  | 10 | 78 |
| 7  | 6  | 82 |
| 6  | 12 | 90 |
| 8  | 9  | 88 |
| 4  | 5  | 65 |
| 9  | 11 | 95 |
| 5  | 7  | 74 |

1. Use the least squares Python function to solve the GLM, and print the resulting equation/model (e.g.: $y = B_2x_2 + B_1x_1 + B_0$).
2. Print the predicted score for 6 hours of self-study per week and 8 completed assignments.
3. Print the SSE value.

```python
hours  = np.array([3, 5, 7, 6, 8, 4, 9, 5], dtype=float)
week   = np.array([8,10, 6,12, 9, 5,11, 7], dtype=float)
score  = np.array([68,78,82,90,88,65,95,74], dtype=float)

# Design matrix: [x1, x2, 1]
X = np.column_stack([hours, week, np.ones(len(hours))])

beta, _, _, _ = np.linalg.lstsq(X, score, rcond=None)
print(f"Model: y = {beta[0]:.4f}·x1 + {beta[1]:.4f}·x2 + {beta[2]:.4f}")

# 2. Predict for x1=6, x2=8
x_new = np.array([6, 8, 1])
y_pred = x_new @ beta
print(f"Predicted score (x1=6, x2=8): {y_pred:.4f}")

# 3. SSE
y_hat = X @ beta
SSE = np.sum((score - y_hat)**2)
print(f"SSE: {SSE:.4f}")
```

1. The columns of Q form a basis for R^n.
2. The determinant of Q is always 1.
3. Q^T = -Q.
4. The eigenvalues of Q are always real.
5. None of the above.

2. The determinant of Q is always 1

$Q^TQ = I$

If we take the determinant on both sides:

$$\det(Q^TQ) = \det(I) = 1$$

We use: $\det(AB) = \det(A)\det(B)$ and $\det(Q^T) = \det(Q)$

$$\det(Q^T)\det(Q) = \det(Q)^2 = 1$$

$$ \det(Q) = \pm 1 \implies |\det(Q)| = 1$$

The geometric intuition for projecting a vector b:

Vector b, is probably not in our subspace, therefore we find the closest point that does.

$$\hat{\mathbf{y}} = X\hat{\boldsymbol{\beta}}, \qquad \mathbf{e} = \mathbf{y} - \hat{\mathbf{y}}$$

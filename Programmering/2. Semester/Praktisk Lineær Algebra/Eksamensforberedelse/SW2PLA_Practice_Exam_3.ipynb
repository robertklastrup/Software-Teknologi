## Question 1 — Vector Projection & Angle

Given vectors:

$$u = \begin{bmatrix} 4 \\ 2 \\ 4 \end{bmatrix}, \quad v = \begin{bmatrix} 1 \\ 0 \\ 2 \end{bmatrix}$$

1. Compute and print the **projection** $u_\parallel$ of $u$ onto $v$.
2. Compute and print the **orthogonal complement** $u_\perp = u - u_\parallel$.
3. Verify orthogonality: print the dot product $u_\perp \cdot v$ (should be 0).
4. Compute and print the **angle** between $u$ and $v$ in degrees.


````python
# Question 1
u = np.array([4, 2, 4], dtype=float)
v = np.array([1, 0, 2], dtype=float)

# 1. Projection of u onto v
# your code
dot_product_u_v = np.dot(u, v)
u_parallel = (dot_product_u_v / np.dot(v, v)) * v
print("Projection of u_parallel = ", u_parallel)


# 2. Orthogonal complement
# your code
u_perp = u - u_parallel
print("Orthogonal complement, u_perp = ", u_perp)

# 3. Verify orthogonality
# your code
print("u_perp · v = ", np.round(np.dot(u_perp, v), 4), "(should be ~0)")


# 4. Angle in degrees
# your code
norm_u = np.linalg.norm(u)
norm_v = np.linalg.norm(v)
angle_rad = np.arccos(np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v)))
print("Angle: ", np.degrees(angle_rad), "°")
````

## Question 2 — Transformation Matrices

A triangle has vertices at the following points (in homogeneous coordinates):

```python
triangle = np.array([[0, 0, 1],
                     [2, 0, 1],
                     [1, 2, 1]], dtype=float)
```

1. Construct a **rotation matrix** (homogeneous, $3\times3$) that rotates **60°** counterclockwise. Print the matrix.
2. Apply the rotation to the triangle and print the transformed vertices.
3. Construct a **translation matrix** that moves all points by $(+3, +1)$. Print the matrix.
4. Apply the translation to the **original** triangle and print the translated vertices.
5. Combine: rotate first, then translate. Print the final vertices.


````python
# Question 2
triangle = np.array([[0, 0, 1],
                     [2, 0, 1],
                     [1, 2, 1]], dtype=float)

# 1. Rotation matrix (60°)
# your code
theta = np.radians(60)
R = np.array([
    [np.cos(theta), -np.sin(theta), 0],
    [np.sin(theta),  np.cos(theta), 0],
    [0,              0,             1]
    ])
print("Rotation matrix (60°):\n", R)

# 2. Rotated triangle
# your code
rotated = triangle @ R
print ("Triangle rotated 60°: \n",rotated)

# 3. Translation matrix (+3, +1)
# your code
T = np.array([[1, 0, 0],
              [0, 1, 0],
              [3, 1, 1]])
print ("Translation matrix: \n", T)

# 4. Translated triangle (from original)
# your code
translated = triangle @ T
print ("Translated triangle: \n", translated)


# 5. Rotate then translate (combined)
# your code
print ("First rotated:\n",rotated, "\n Translated: \n", rotated @ T)
````

Consider the following three systems of equations:

**System A:**
$$\begin{cases} x + 2y + 3z = 6 \\ 2x + 5y + 2z = 4 \\ x + 3y - z = -2 \end{cases}$$

**System B:**
$$\begin{cases} x + 2y = 3 \\ 2x + 4y = 6 \end{cases}$$

**System C:**
$$\begin{cases} x + y = 5 \\ 2x + 2y = 3 \end{cases}$$

For **each** system:
1. Compute and print the **RREF** of the augmented matrix $[A|b]$.
2. Print `rank(A)` and `rank([A|b])`.
3. State how many solutions the system has and **why** (use rank reasoning).


### Answer — Number of solutions reasoning

*(Write your answers here for each system)*


````python
# Question 3 — System A
print ("System A")
A_sys = np.array([[1, 2, 3],
                  [2, 5, 2],
                  [1, 3, -1]], dtype=float)
b_sys = np.array([[6], [4], [-2]], dtype=float)

# RREF and rank analysis
# your code
Ab = np.column_stack([A_sys, b_sys])
rref, pivots = Matrix(Ab).rref()
print(f"\nRREF of [A|b]:")
print(rref)
rank_A  = np.linalg.matrix_rank(A_sys)
rank_Ab = np.linalg.matrix_rank(np.column_stack([A_sys, b_sys]))

print(f"Rank A:    {rank_A}")
print(f"Rank [A|b]: {rank_Ab}")

if rank_A == rank_Ab == A_sys.shape[1]:
    x = np.linalg.solve(A_sys, b_sys)
    print(f"Unique solution: {x}\n")
elif rank_A == rank_Ab:
    print("Infinitely many solutions, since the rank < n\n")
else:
    print("No solution\n")

#System B:
print ("System B \n")

A_sysB = np.array([[1,2],
                   [2, 4]])
b_sysB = np.array([[3], [6]])

Ab_B = np.column_stack([A_sysB, b_sysB])
rref, pivots = Matrix(Ab_B).rref()
print(f"\nRREF of [A|b]:")
print(rref)
rank_A_B  = np.linalg.matrix_rank(A_sysB)
rank_Ab_B = np.linalg.matrix_rank(np.column_stack([A_sysB, b_sysB]))

print(f"Rank A:    {rank_A_B}")
print(f"Rank [A|b]: {rank_Ab_B}")

if rank_A_B == rank_Ab_B == A_sysB.shape[1]:
    xB = np.linalg.solve(A_sysB, b_sysB)
    print(f"Unique solution: {xB}\n")
elif rank_A_B == rank_Ab_B:
    print("Infinitely many solutions, since the rank < n\n")
else:
    print("No solution\n")

#System C:
print ("System C")

A_sysC = np.array([[1, 1],
                   [2, 2]])
b_sysC = np.array ([[5], [3]])

Ab_C = np.column_stack([A_sysC, b_sysC])
rref, pivots = Matrix(Ab_C).rref()
print(f"\nRREF of [A|b]:")
print(rref)
rank_A_C  = np.linalg.matrix_rank(A_sysC)
rank_Ab_C = np.linalg.matrix_rank(np.column_stack([A_sysC, b_sysC]))

print(f"Rank A:    {rank_A_C}")
print(f"Rank [A|b]: {rank_Ab_C}")

if rank_A_C == rank_Ab_C == A_sysC.shape[1]:
    xC = np.linalg.solve(A_sysC, b_sysC)
    print(f"Unique solution: {xC} \n")
elif rank_A_C == rank_Ab_C:
    print("Infinitely many solutions, since the rank < n \n")
else:
    print("No solution \n")
````

```python
              [1.5, 2.0, 4.5]], dtype=float)
```

1. **Mean-center** the data (subtract column means). Print the column means.
2. Compute and print the **covariance matrix** $C = \frac{X^T X}{n-1}$.
3. Verify that $C$ is **symmetric**.
4. Compute and print the **Pearson correlation coefficient** between feature 1 and feature 3 (manually using the formula, not `np.corrcoef`).


````python
# Question 4
X = np.array([[2.5, 3.1, 8.0],
              [1.0, 1.5, 3.5],
              [3.0, 3.8, 9.5],
              [2.0, 2.5, 6.0],
              [4.0, 4.5, 12.0],
              [1.5, 2.0, 4.5]], dtype=float)

# 1. Mean-center
# your code
mean = X - np.mean(X, axis=0, keepdims=True)
print("Column means:", np.mean(X, axis=0))
print("Mean-centered data:\n", np.round(mean, 4))


# 2. Covariance matrix
# your code
n = mean.shape[0]
C = mean.T @ mean / (n - 1)
print(f"\nCovariance matrix C:")
print(np.round(C, 4))

# 3. Verify symmetric
# your code
if np.allclose(C, C.T):
    print("\nC er symmetrisk eftersom C = C.T\n")
else:
    print("\nC er ikke symmetrisk\n")

# 4. Pearson correlation (manual) between feature 0 and feature 2
# your code
corr = C[0,2] / np.sqrt(C[0,0] * C[2,2])
print(f"\nPearson correlation: {corr:.4f}")

````

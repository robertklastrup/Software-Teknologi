## Question 1 — Vector Projection & Angle

Given vectors:

$$u = \begin{bmatrix} 4 \\ 2 \\ 4 \end{bmatrix}, \quad v = \begin{bmatrix} 1 \\ 0 \\ 2 \end{bmatrix}$$

1. Compute and print the **projection** $u_\parallel$ of $u$ onto $v$.
2. Compute and print the **orthogonal complement** $u_\perp = u - u_\parallel$.
3. Verify orthogonality: print the dot product $u_\perp \cdot v$ (should be 0).
4. Compute and print the **angle** between $u$ and $v$ in degrees.


````python
# Question 1
u = np.array([4, 2, 4], dtype=float)
v = np.array([1, 0, 2], dtype=float)

# 1. Projection of u onto v
# your code
dot_product_u_v = np.dot(u, v)
u_parallel = (dot_product_u_v / np.dot(v, v)) * v
print("Projection of u_parallel = ", u_parallel)


# 2. Orthogonal complement
# your code
u_perp = u - u_parallel
print("Orthogonal complement, u_perp = ", u_perp)

# 3. Verify orthogonality
# your code
print("u_perp · v = ", np.round(np.dot(u_perp, v), 4), "(should be ~0)")


# 4. Angle in degrees
# your code
norm_u = np.linalg.norm(u)
norm_v = np.linalg.norm(v)
angle_rad = np.arccos(np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v)))
print("Angle: ", np.degrees(angle_rad), "°")
````

## Question 2 — Transformation Matrices

A triangle has vertices at the following points (in homogeneous coordinates):

```python
triangle = np.array([[0, 0, 1],
                     [2, 0, 1],
                     [1, 2, 1]], dtype=float)
```

1. Construct a **rotation matrix** (homogeneous, $3\times3$) that rotates **60°** counterclockwise. Print the matrix.
2. Apply the rotation to the triangle and print the transformed vertices.
3. Construct a **translation matrix** that moves all points by $(+3, +1)$. Print the matrix.
4. Apply the translation to the **original** triangle and print the translated vertices.
5. Combine: rotate first, then translate. Print the final vertices.


````python
# Question 2
triangle = np.array([[0, 0, 1],
                     [2, 0, 1],
                     [1, 2, 1]], dtype=float)

# 1. Rotation matrix (60°)
# your code
theta = np.radians(60)
R = np.array([
    [np.cos(theta), -np.sin(theta), 0],
    [np.sin(theta),  np.cos(theta), 0],
    [0,              0,             1]
    ])
print("Rotation matrix (60°):\n", R)

# 2. Rotated triangle
# your code
rotated = triangle @ R
print ("Triangle rotated 60°: \n",rotated)

# 3. Translation matrix (+3, +1)
# your code
T = np.array([[1, 0, 0],
              [0, 1, 0],
              [3, 1, 1]])
print ("Translation matrix: \n", T)

# 4. Translated triangle (from original)
# your code
translated = triangle @ T
print ("Translated triangle: \n", translated)


# 5. Rotate then translate (combined)
# your code
print ("First rotated:\n",rotated, "\n Translated: \n", rotated @ T)
````

Given matrix:

$$A = \begin{bmatrix} 3 & 1 \\ 1 & 3 \end{bmatrix}$$

2. Verify your result using `np.linalg.eig`.
3. Verify that $\text{tr}(A) = \sum \lambda_i$ and $\det(A) = \prod \lambda_i$.
4. Diagonalize $A$ — print $V$, $D$, and $V^{-1}$, and verify reconstruction.


````python
# Question 5
A = np.array([[3, 1],
              [1, 3]], dtype=float)

# 1. Eigenvalues from characteristic polynomial
# your code
# 1. Eigenvalues fra karakteristisk polynomium: λ² - tr(A)λ + det(A) = 0
coeffs = [1, -np.trace(A), np.linalg.det(A)]  # [1, -6, 8]
eigenvalues_poly = np.roots(coeffs)
print("Eigenvalues (fra polynomium):", eigenvalues_poly)

# 2. Verify with np.linalg.eig
# your code
eigenvalues_eig, V = np.linalg.eig(A)
print("Eigenvalues (np.linalg.eig):", eigenvalues_eig)

# 3. Verify tr(A) = sum(λ) and det(A) = prod(λ)
# your code
print("tr(A) =", np.trace(A), "| sum(λ) =", np.sum(eigenvalues_eig))
print("det(A) =", np.linalg.det(A), "| prod(λ) =", np.prod(eigenvalues_eig))

# 4. Diagonalize and verify reconstruction
# your code
D = np.diag(eigenvalues_eig)
V_inv = np.linalg.inv(V)
print("V:\n", V)
print("D:\n", D)
print("V_inv:\n", V_inv)
print("Rekonstruktion V @ D @ V_inv:\n", np.round(V @ D @ V_inv, 6))
print("Original A:\n", A)
````

A dataset records three variables for 8 students: hours studied, hours slept, and exam score.

| Hours studied | Hours slept | Exam score |
|:---:|:---:|:---:|
| 5 | 7 | 72 |
| 8 | 6 | 85 |
| 3 | 8 | 60 |
| 9 | 7 | 91 |
| 6 | 5 | 70 |
| 7 | 8 | 83 |
| 4 | 6 | 65 |
| 10 | 7 | 95 |

Fit the model: $y = \beta_0 + \beta_1 \cdot \text{studied} + \beta_2 \cdot \text{slept}$

1. Build the **design matrix** $X$ (columns: studied, slept, ones). Print it.
2. Solve using **least squares**. Print the model equation.
3. Predict the score for a student who studied **7.5 hours** and slept **7 hours**.
4. Compute and print **SSE** and **R²**.


````python
# Question 6
studied = np.array([5, 8, 3, 9, 6, 7, 4, 10], dtype=float)
slept   = np.array([7, 6, 8, 7, 5, 8, 6,  7], dtype=float)
scores  = np.array([72, 85, 60, 91, 70, 83, 65, 95], dtype=float)

# 1. Design matrix
# your code
X = np.column_stack([studied, slept, np.ones(len(studied))])
print("Design matrix:\n", X)

# 2. Solve least squares and print model equation
# your code
result = np.linalg.lstsq(X, scores, rcond=None)
beta1, beta2, beta0 = result[0]
print(f"Model: y = {beta0:.4f} + {beta1:.4f}*studied + {beta2:.4f}*slept")

# 3. Predict for studied=7.5, slept=7
# your code
y_pred = beta0 + beta1 * 7.5 + beta2 * 7
print(f"Predicted score: {y_pred:.2f}")

# 4. SSE and R²
# your code
scores_predicted = X @ result[0]
SSE = np.sum((scores - scores_predicted)**2)
SS_tot = np.sum((scores - np.mean(scores))**2)
R2 = 1 - SSE / SS_tot
print(f"SSE: {SSE:.4f}")
print(f"R²: {R2:.4f}")
````

## Question 7 — QR-Based Inverse & Inverse Properties

Given matrices:

$$A = \begin{bmatrix} 2 & 1 & 1 \\ 0 & 3 & 1 \\ 1 & 1 & 4 \end{bmatrix}, \quad B = \begin{bmatrix} 1 & 2 & 0 \\ 0 & 1 & 1 \\ 2 & 0 & 1 \end{bmatrix}$$

1. Compute $A^{-1}$ using **QR decomposition** (formula: $A^{-1} = R^{-1}Q^T$). Print the result.
2. Verify it matches `np.linalg.inv(A)`.
3. Compute $(AB)^{-1}$ and $B^{-1}A^{-1}$. Are they equal? Print a boolean.
4. Compute $A^{-1}B^{-1}$. Is it equal to $(AB)^{-1}$? Print a boolean and explain.


### Answer to Question 7.4

*(Explain in your own words why the order matters)*


````python
# Question 7
A = np.array([[2, 1, 1],
              [0, 3, 1],
              [1, 1, 4]], dtype=float)
B = np.array([[1, 2, 0],
              [0, 1, 1],
              [2, 0, 1]], dtype=float)

# 1. QR-based inverse of A
# your code
# A⁻¹ via QR: A⁻¹ = R⁻¹ @ Q.T
Q, R = np.linalg.qr(A)
A_inv_qr = np.linalg.inv(R) @ Q.T
print("A⁻¹ via QR:\n", np.round(A_inv_qr, 6))

# 2. Verify matches np.linalg.inv(A)
# your code
A_inv = np.linalg.inv(A)
print("Matcher np.linalg.inv:", np.allclose(A_inv_qr, A_inv))

# 3. (AB)⁻¹ vs B⁻¹A⁻¹
# your code
AB_inv = np.linalg.inv(A @ B)
B_inv_A_inv = np.linalg.inv(B) @ np.linalg.inv(A)
print("(AB)⁻¹ == B⁻¹A⁻¹:", np.allclose(AB_inv, B_inv_A_inv))

# 4. A⁻¹B⁻¹ — is it equal to (AB)⁻¹?
# your code
A_inv_B_inv = np.linalg.inv(A) @ np.linalg.inv(B)
print("A⁻¹B⁻¹ == (AB)⁻¹:", np.allclose(A_inv_B_inv, AB_inv))

````

## Why the order matters:

Matrix multiplication is not commutative, meaning that $AB \neq BA$ in general.
This is why the order reverses when taking the inverse of a product.

To see why, $(AB)^{-1}$ must satisfy $(AB)(AB)^{-1} = I$.
If we try $A^{-1}B^{-1}$, we get:

$$AB \cdot A^{-1}B^{-1}$$

which does not simplify to $I$, because the middle terms $B A^{-1}$ cannot be cancelled.

However, with $B^{-1}A^{-1}$ we get:

$$AB \cdot B^{-1}A^{-1} = A(BB^{-1})A^{-1} = AA^{-1} = I$$

which confirms the correct rule:

$$(AB)^{-1} = B^{-1}A^{-1}$$

Therefore $A^{-1}B^{-1} \neq (AB)^{-1}$ in general, unless $A$ and $B$ happen to commute.

## Question 9 — GED / LDA

The following data represents two classes:

```python
class1 = np.array([[1.2, 0.8], [1.5, 1.1], [0.9, 0.7],
                   [1.3, 0.9], [1.1, 1.0], [1.4, 0.8]])
class2 = np.array([[3.5, 2.8], [3.8, 3.1], [3.2, 2.6],
                   [3.6, 2.9], [3.4, 3.0], [3.9, 2.8]])
```

1. Compute and print **covB** (between-class covariance).
2. Compute and print **covW** (within-class covariance).
3. Solve the **GED** using `scipy.linalg.eig(covB, covW)`. Print the eigenvalues.
4. Project both classes onto the **best discriminant axis** (first eigenvector). Print the projected values.


````python
# Question 9
class1 = np.array([[1.2, 0.8], [1.5, 1.1], [0.9, 0.7],
                   [1.3, 0.9], [1.1, 1.0], [1.4, 0.8]])
class2 = np.array([[3.5, 2.8], [3.8, 3.1], [3.2, 2.6],
                   [3.6, 2.9], [3.4, 3.0], [3.9, 2.8]])

# 1. covB (between-class covariance)
mean1 = np.mean(class1, axis=0)
mean2 = np.mean(class2, axis=0)
mean_total = (mean1 + mean2) / 2

covB = (np.outer(mean1 - mean_total, mean1 - mean_total) +
        np.outer(mean2 - mean_total, mean2 - mean_total))
print("covB:\n", covB)

# 2. covW (within-class covariance)
covW = np.cov(class1.T, ddof=1) + np.cov(class2.T, ddof=1)
print("covW:\n", covW)

# 3. GED
eigenvalues, eigenvectors = scipy.linalg.eig(covB, covW)
print("Eigenvalues:", eigenvalues.real)

# 4. Project onto best discriminant axis (first eigenvector)
best_axis = eigenvectors[:, np.argmax(eigenvalues.real)].real
proj1 = class1 @ best_axis
proj2 = class2 @ best_axis
print("Class 1 projected:", proj1)
print("Class 2 projected:", proj2)
````

````python
# Question 11 — Part A (verify Cauchy-Schwarz numerically)
u = np.array([1, 5, 7], dtype=float)
v = np.array([2, 5, 9], dtype=float)

# Verify |u·v| ≤ ||u|| * ||v||
# your code

lhs = abs(np.dot(u, v))
rhs = np.linalg.norm(u) * np.linalg.norm(v)
print(f"|u·v| = {lhs:.4f}")
print(f"||u||·||v|| = {rhs:.4f}")
print(f"Cauchy-Schwarz holder: {lhs <= rhs}")

````

### 1.1 Komplement og "mindst én"

**Bruges når:** Opgaven spørger om sandsynligheden for at **mindst én/ét** ud af flere (uafhængige) hændelser indtræffer — fx "mindst én sensor opdager fejlen", "mindst ét angreb opdages", "systemet virker hvis mindst én komponent virker" (parallelt system).

**Virker uanset om sandsynlighederne er ens eller forskellige** — det er netop pointen ved at bruge komplementet: $P(\text{mindst én})=1-P(\text{ingen})$, og $P(\text{ingen})$ er blot produktet af alle komplementerne, uanset om de er ens.

**Bruges IKKE når:** opgaven spørger om **præcis k** ud af n — se 1.7–1.9 i stedet.

```python
# === IMPORTS ===
# (ingen eksterne pakker nødvendige)

# === INDSÆT DINE VÆRDIER HER ===
# Sandsynligheden for at HVER hændelse selv indtræffer (ikke komplementet!)
P_haendelser = [0.9, 0.85, 0.8]   # fx P(sensor A opdager), P(sensor B opdager), P(sensor C opdager)

# === BEREGNING ===
P_ingen = 1.0
for p in P_haendelser:
    P_ingen *= (1 - p)

P_mindst_en = 1 - P_ingen

# === RESULTATER ===
print(f"P(ingen af hændelserne indtræffer)   = {P_ingen:.4f}")
print(f"P(mindst én af hændelserne indtræffer) = {P_mindst_en:.4f}")
```

### 1.2 Betinget sandsynlighed og union

**Bruges når:** Ordet **"givet"** optræder i opgaven (betinget sandsynlighed), eller du skal finde sandsynligheden for at **A eller B** sker (union / inklusion-eksklusion).

$$P(A\mid B)=\frac{P(A\cap B)}{P(B)} \qquad\qquad P(A\cup B)=P(A)+P(B)-P(A\cap B)$$

Hvis $A$ og $B$ er **uafhængige**, kan $P(A\cap B)$ erstattes med $P(A)\cdot P(B)$ i begge formler.

```python
# === IMPORTS ===
# (ingen eksterne pakker nødvendige)

# === INDSÆT DINE VÆRDIER HER ===
P_A = 0.42          # P(A)
P_B = 0.72          # P(B)
P_A_and_B = 0.30    # P(A ∩ B) — sæt til None hvis A,B er uafhængige og du vil udlede den
uafhaengige = False  # sæt til True hvis A og B er uafhængige (så udregnes P(A∩B) = P(A)*P(B))

# === BEREGNING ===
if uafhaengige:
    P_A_and_B = P_A * P_B

P_A_given_B = P_A_and_B / P_B
P_B_given_A = P_A_and_B / P_A
P_union = P_A + P_B - P_A_and_B

# === RESULTATER ===
print(f"P(A ∩ B)   = {P_A_and_B:.4f}")
print(f"P(A | B)   = {P_A_given_B:.4f}")
print(f"P(B | A)   = {P_B_given_A:.4f}")
print(f"P(A ∪ B)   = {P_union:.4f}")
```

### 1.3 Total sandsynlighed

**Bruges når:** Problemet kan opdeles i **gensidigt udelukkende scenarier** (en partition) — fx maskintype, årstid, sygdomsstatus — og du kender $P(A\mid B_i)$ og $P(B_i)$ for hvert scenarie, men skal finde den samlede/ubetingede $P(A)$.

$$P(A)=\sum_i P(A\mid B_i)\cdot P(B_i)$$

**Signalord:** "en tilfældigt udvalgt ... hvad er sandsynligheden for ..." hvor tingen kommer fra én ud af flere kilder/grupper.

```python
# === IMPORTS ===
# (ingen eksterne pakker nødvendige)

# === INDSÆT DINE VÆRDIER HER ===
# Ét tal pr. scenarie/partition (fx maskine, årstid, gruppe)
navne          = ["Forår", "Sommer", "Efterår", "Vinter"]
P_A_givet_Bi   = [0.46,    0.63,     0.32,      0.12]   # P(A | scenarie_i)
P_Bi           = [0.25,    0.25,     0.25,      0.25]   # P(scenarie_i) — skal summere til 1

# === BEREGNING ===
assert abs(sum(P_Bi) - 1) < 1e-9, "P(B_i) summerer ikke til 1 — tjek dine partition-sandsynligheder!"

P_A = sum(p_a * p_b for p_a, p_b in zip(P_A_givet_Bi, P_Bi))

# === RESULTATER ===
for navn, p_a, p_b in zip(navne, P_A_givet_Bi, P_Bi):
    print(f"  {navn:10s}: P(A|{navn})·P({navn}) = {p_a:.3f} · {p_b:.3f} = {p_a*p_b:.4f}")
print(f"\nP(A) (total sandsynlighed) = {P_A:.4f}")
```

### 1.4 Bayes' sætning

**Bruges når:** Du kender $P(A\mid B_i)$ (fx testens træfsikkerhed givet sygdom), men skal finde $P(B_i\mid A)$ (fx sandsynlighed for sygdom givet positiv test) — betingelsen skal **vendes om**.

$$P(B_i\mid A)=\frac{P(A\mid B_i)\,P(B_i)}{\sum_j P(A\mid B_j)\,P(B_j)}$$

Nævneren er præcis total-sandsynlighedsformlen fra 1.3 — så dette snippet beregner også $P(A)$ undervejs.

**Tip:** Vil du hellere se beregningen udført **trin for trin** (træ-metoden) i stedet for som ét langt Bayes-udtryk? Se Eksamensbogen, Del I §1.4.1.

```python
# === IMPORTS ===
# (ingen eksterne pakker nødvendige)

# === INDSÆT DINE VÆRDIER HER ===
navne        = ["M1", "M2"]         # navne på scenarierne/grupperne B_i
P_A_givet_Bi = [0.03, 0.05]         # P(A | B_i), fx P(defekt | maskine_i)
P_Bi         = [0.6,  0.4]          # P(B_i), fx P(maskine_i) — skal summere til 1
maal_index   = 1                    # hvilket scenarie du vil finde P(B_i | A) for (0-indekseret)

# === BEREGNING ===
assert abs(sum(P_Bi) - 1) < 1e-9, "P(B_i) summerer ikke til 1!"

P_A = sum(p_a * p_b for p_a, p_b in zip(P_A_givet_Bi, P_Bi))
P_Bmaal_and_A = P_A_givet_Bi[maal_index] * P_Bi[maal_index]
P_Bmaal_givet_A = P_Bmaal_and_A / P_A

# === RESULTATER ===
print(f"P(A) (nævner, total sandsynlighed) = {P_A:.4f}")
print(f"P({navne[maal_index]} ∩ A) = {P_Bmaal_and_A:.4f}")
print(f"P({navne[maal_index]} | A) = {P_Bmaal_givet_A:.4f}   <-- Bayes' svar")
```

### 1.5 Kædereglen (uden tilbagelægning)

**Bruges når:** Der trækkes **successivt uden tilbagelægning**, så sandsynligheden ændrer sig for hvert træk (fx "alle 3 udtrukne er defekte", "de første 2 poser er grønne" med begrænset antal).

$$P(A_1\cap A_2\cap\cdots\cap A_n)=P(A_1)\cdot P(A_2\mid A_1)\cdot P(A_3\mid A_1,A_2)\cdots$$

Ved uden-tilbagelægning fra en population med $N$ i alt og $K$ af "den ønskede type", falder både tæller og nævner med 1 for hvert træk.

```python
# === IMPORTS ===
# (ingen eksterne pakker nødvendige)

# === INDSÆT DINE VÆRDIER HER ===
N = 15   # totalt antal i populationen
K = 4    # antal af "den ønskede type" (fx defekte)
k = 3    # antal træk du vil have alle af typen "K" blandt (fx "alle 3 udtrukne er defekte")

# === BEREGNING ===
P = 1.0
tekst_led = []
for i in range(k):
    p_i = (K - i) / (N - i)
    P *= p_i
    tekst_led.append(f"({K-i})/({N-i})")

# === RESULTATER ===
print("P =", " · ".join(tekst_led), "=", round(P, 5))
```

### 1.6 Kombinatorik: permutation vs. kombination

**Bruges når:** Opgaven spørger "hvor mange måder kan ... udvælges/fordeles". Stil dig altid disse to spørgsmål først:

| | Med tilbagelægning | Uden tilbagelægning | Kendetegn |
|---|---|---|---|
| **Ordnet** | $n^k$ | $n!/(n-k)!$ | Positioner/roller er forskellige (1. plads, 2. plads, ...) |
| **Uordnet** | $\binom{n+k-1}{k}$ | $\binom{n}{k}$ | Man vælger blot en gruppe/mængde, rækkefølgen er ligegyldig |

```python
# === IMPORTS ===
import math

# === INDSÆT DINE VÆRDIER HER ===
n = 12    # antal elementer at vælge/fordele imellem
k = 4     # antal der skal vælges/fordeles

# === BEREGNING ===
ordnet_uden      = math.perm(n, k)               # n! / (n-k)!
ordnet_med       = n ** k                         # n^k
uordnet_uden     = math.comb(n, k)                # n! / (k!(n-k)!)
uordnet_med      = math.comb(n + k - 1, k)        # multimængde

# === RESULTATER ===
print(f"Ordnet,  uden tilbagelægning (permutation) : P(n,k) = {ordnet_uden}")
print(f"Ordnet,  med tilbagelægning                : n^k    = {ordnet_med}")
print(f"Uordnet, uden tilbagelægning (kombination)  : C(n,k) = {uordnet_uden}")
print(f"Uordnet, med tilbagelægning (multimængde)   : C(n+k-1,k) = {uordnet_med}")
print()
print(">> Vælg den relevante af de fire ovenfor ud fra opgavens ordlyd (se tabellen).")
```

### 1.7 Hypergeometrisk fordeling

**Bruges når:** Du udtager $n$ elementer **uden tilbagelægning** fra en population på $N$ elementer, hvoraf $K$ er "af den ønskede type", og vil vide sandsynligheden for **præcis $k$** af dem er af den type.

$$P(X=k)=\frac{\binom{K}{k}\binom{N-K}{n-k}}{\binom{N}{n}}$$

**Forveksl ikke med binomialfordelingen (1.8)** — her er der IKKE tilbagelægning, så sandsynligheden ændrer sig undervejs (det er derfor formlen bruger kombinationer af hele populationen, ikke $p^k(1-p)^{n-k}$).

```python
# === IMPORTS ===
from scipy.stats import hypergeom

# === INDSÆT DINE VÆRDIER HER ===
N = 15   # population i alt
K = 4    # antal "af den ønskede type" i populationen (fx defekte)
n = 3    # antal udtrukket (uden tilbagelægning)
k = 1    # antal "af den ønskede type" blandt de udtrukne, du spørger om

# === BEREGNING ===
rv = hypergeom(N, K, n)
P_eksakt   = rv.pmf(k)                 # P(X = k)
P_mindst_k = 1 - rv.cdf(k - 1)         # P(X >= k)
P_hoejst_k = rv.cdf(k)                 # P(X <= k)

# === RESULTATER ===
print(f"P(X = {k})     = {P_eksakt:.4f}")
print(f"P(X >= {k})    = {P_mindst_k:.4f}")
print(f"P(X <= {k})    = {P_hoejst_k:.4f}")
```

### 1.8 Binomialfordeling

**Bruges når:** $n$ **identiske, uafhængige** Bernoulli-forsøg med **samme** succes-sandsynlighed $p$ i hvert forsøg (fx "10% af enhederne er defekte, find P(præcis 3 defekte ud af 50)").

$$P(X=k)=\binom{n}{k}p^k(1-p)^{n-k}, \qquad E[X]=np, \quad Var(X)=np(1-p)$$

```python
# === IMPORTS ===
from scipy.stats import binom

# === INDSÆT DINE VÆRDIER HER ===
n = 50     # antal uafhængige forsøg
p = 0.1    # succes-sandsynlighed (samme for hvert forsøg!)
k = 15     # antal succeser du spørger om

# === BEREGNING ===
P_eksakt   = binom.pmf(k, n, p)
P_hoejst_k = binom.cdf(k, n, p)
P_mindst_k = 1 - binom.cdf(k - 1, n, p)
E_X = n * p
Var_X = n * p * (1 - p)

# === RESULTATER ===
print(f"P(X = {k})   = {P_eksakt:.6f}")
print(f"P(X <= {k})  = {P_hoejst_k:.6f}")
print(f"P(X >= {k})  = {P_mindst_k:.6f}")
print(f"E[X]   = {E_X}")
print(f"Var(X) = {Var_X}")
```

### 1.9 Forskellige sandsynligheder — direkte multiplikation (IKKE binomial)

**Bruges når:** Flere uafhængige hændelser med **forskellige** sandsynligheder (fx tre sensorer med hver deres træfsikkerhed), og du skal finde $P(\text{præcis } k \text{ af dem indtræffer})$.

**Faldgrube:** Binomialformlen (1.8) gælder KUN når alle forsøg har samme $p$. Her skal du i stedet summere sandsynligheden for hvert af de $\binom{n}{k}$ disjunkte scenarier direkte. Dette snippet gør det automatisk med `itertools.combinations`, uanset hvor mange hændelser du har.

```python
# === IMPORTS ===
from itertools import combinations

# === INDSÆT DINE VÆRDIER HER ===
# Sandsynligheden for at HVER hændelse selv indtræffer (ikke komplementet!)
P_haendelser = [0.9, 0.85, 0.8]   # fx P(sensor A), P(sensor B), P(sensor C) — kan være vilkårligt mange
k = 2                              # antal, der skal indtræffe præcis

# === BEREGNING ===
n = len(P_haendelser)
idx = range(n)
P_praecis_k = 0.0
for succes_idx in combinations(idx, k):
    p = 1.0
    for i in idx:
        p *= P_haendelser[i] if i in succes_idx else (1 - P_haendelser[i])
    P_praecis_k += p

# Bonus: hele fordelingen (P(X=0), P(X=1), ..., P(X=n)) på én gang
fordeling = []
for j in range(n + 1):
    P_j = sum(
        __import__("math").prod(
            P_haendelser[i] if i in s else (1 - P_haendelser[i]) for i in idx
        )
        for s in combinations(idx, j)
    )
    fordeling.append(P_j)

# === RESULTATER ===
print(f"P(præcis {k} indtræffer) = {P_praecis_k:.4f}")
print()
print("Hele fordelingen (sanity-check: bør summere til 1):")
for j, P_j in enumerate(fordeling):
    print(f"  P(X = {j}) = {P_j:.4f}")
print(f"  Sum = {sum(fordeling):.4f}")
```

### 1.10 Serie- og parallelsystemer (reliability)

**Bruges når:** Opgaven beskriver et **system af komponenter** (sensorer, generatorer, servere) og spørger om sandsynligheden for at **hele systemet** virker.

- **Parallelt** ("mindst én komponent skal virke", "redundant", "backup"): $P(\text{virker})=1-\prod(1-P_i)$ — samme som 1.1.
- **Serie** ("alle komponenter skal virke"): $P(\text{virker})=\prod P_i$.

```python
# === IMPORTS ===
# (ingen eksterne pakker nødvendige)

# === INDSÆT DINE VÆRDIER HER ===
P_komponenter = [0.9, 0.95]   # sandsynlighed for at HVER komponent virker
system_type = "parallel"       # "parallel" (mindst én skal virke) eller "serie" (alle skal virke)

# === BEREGNING ===
if system_type == "serie":
    P_system = 1.0
    for p in P_komponenter:
        P_system *= p
else:  # parallel
    P_system = 1.0
    for p in P_komponenter:
        P_system *= (1 - p)
    P_system = 1 - P_system

# === RESULTATER ===
print(f"System-type: {system_type}")
print(f"P(systemet virker) = {P_system:.4f}")
```

### 2.1 Diskret PMF: gyldighed, E[X], Var(X)

**Bruges når:** $X$ er givet ved konkrete sandsynligheder $P(X=x)$ for et **tælleligt** antal værdier (tabel, liste, eller figur uden ukendte konstanter — se 2.3 hvis der er en ukendt konstant).

$$\sum_x f_X(x)=1 \qquad E[X]=\sum_x x\,f_X(x) \qquad Var(X)=E[X^2]-E[X]^2$$

**Diskret = SUM, aldrig integral.**

```python
# === IMPORTS ===
# (ingen eksterne pakker nødvendige)

# === INDSÆT DINE VÆRDIER HER ===
xs = [0, 1, 2, 3]              # de mulige værdier af X
ps = [0.4, 0.35, 0.2, 0.05]    # P(X = x) for hver værdi, i samme rækkefølge

# === BEREGNING ===
assert all(p >= 0 for p in ps), "Alle sandsynligheder skal være >= 0!"
sum_ps = sum(ps)

E_X  = sum(x * p for x, p in zip(xs, ps))
E_X2 = sum(x**2 * p for x, p in zip(xs, ps))
Var_X = E_X2 - E_X**2

# === RESULTATER ===
print(f"Sum af P(X=x) = {sum_ps:.6f}  (skal være 1 for gyldig fordeling)")
print(f"E[X]    = {E_X}")
print(f"E[X^2]  = {E_X2}")
print(f"Var(X)  = {Var_X}")
```

### 2.2 Diskret CDF og invers CDF

**Bruges når:** Du skal opstille $F_X(x)$ som en **trappefunktion** ud fra en diskret PMF, eller finde $F_X^{-1}(u)$ (fx til brug i simulering, se 3.9/3.11).

$$F_X(x)=\sum_{x_i\le x} f_X(x_i) \qquad\qquad F_X^{-1}(u)=\min\{x: F_X(x)\ge u\}$$

```python
# === IMPORTS ===
import numpy as np

# === INDSÆT DINE VÆRDIER HER ===
xs = [0, 1, 2, 3]
ps = [0.4, 0.35, 0.2, 0.05]
x_forespoergsel = 1.5   # find F_X ved denne værdi
u_forespoergsel = 0.6   # find F_X^{-1} ved denne værdi (mellem 0 og 1)

# === BEREGNING ===
xs_sorted = sorted(xs)
ps_sorted = [ps[xs.index(x)] for x in xs_sorted]
cdf_vaerdier = np.cumsum(ps_sorted)

def F(x):
    # Diskret CDF: P(X <= x)
    total = 0.0
    for xi, p in zip(xs_sorted, ps_sorted):
        if xi <= x:
            total += p
    return total

def F_inv(u):
    # Invers CDF: mindste x hvor F(x) >= u
    for xi, cdf_i in zip(xs_sorted, cdf_vaerdier):
        if cdf_i >= u:
            return xi
    return xs_sorted[-1]

# === RESULTATER ===
print("Trappefunktion F_X(x):")
laveste = 0.0
for xi, cdf_i in zip(xs_sorted, cdf_vaerdier):
    print(f"  F_X(x) = {cdf_i:.4f}  for  {xi} <= x < (næste værdi)")
print()
print(f"F_X({x_forespoergsel}) = {F(x_forespoergsel):.4f}")
print(f"F_X^-1({u_forespoergsel}) = {F_inv(u_forespoergsel)}")
```

### 2.3 Diskret PMF med ukendt(e) konstant(er)

**Bruges når:** PMF'en er givet ved en figur/tabel med **én eller flere ukendte** (typisk $a$), som skal bestemmes ud fra normeringsligningen $\sum f_X(x)=1$ — og evt. én ekstra oplysning (fx en given $E[X]$), hvis der er to ukendte.

**Fremgangsmåde:** aflæs værdimængden, udtryk hver sandsynlighed i de ukendte, opstil normeringsligningen (+ en ekstra ligning hvis nødvendigt), og løs.

```python
# === IMPORTS ===
import sympy as sp

# === INDSÆT DINE VÆRDIER HER ===
a, b = sp.symbols('a b', positive=True)   # de ukendte konstanter (slet 'b' hvis du kun har én ukendt)

xs = [0, 1, 2, 3]
# Udtryk HVER sandsynlighed i de ukendte (brug tal direkte hvor værdien er kendt)
ps_symbolsk = [sp.Rational(1, 10), a, b, sp.Rational(1, 10)]

# Ekstra ligning hvis du har 2 ukendte, fx en given E[X] -- sæt til None hvis du kun har 1 ukendt
givet_EX = sp.Rational(16, 10)

# === BEREGNING ===
ligninger = [sp.Eq(sum(ps_symbolsk), 1)]   # normeringsligningen

if givet_EX is not None:
    EX_udtryk = sum(x * p for x, p in zip(xs, ps_symbolsk))
    ligninger.append(sp.Eq(EX_udtryk, givet_EX))

loesning = sp.solve(ligninger, [a, b] if givet_EX is not None else [a])

ps_talvaerdier = [float(p.subs(loesning)) if hasattr(p, "subs") else float(p) for p in ps_symbolsk]

# === RESULTATER ===
print("Løsning:", loesning)
print()
print("PMF med indsatte tal:")
for x, p in zip(xs, ps_talvaerdier):
    print(f"  P(X={x}) = {p:.4f}")
print(f"Sum = {sum(ps_talvaerdier):.4f}  (sanity-check, skal være 1)")
```

### 2.4 Kontinuert PDF: find k, CDF, E[X], Var(X), median

**Bruges når:** $X$ er givet ved en tæthedsfunktion $f_X(x)=k\cdot(\text{udtryk})$ på et interval $[a,b]$, med en ukendt konstant $k$, der skal bestemmes.

$$\int_a^b f_X(x)\,dx=1 \qquad F_X(x)=\int_a^x f_X(u)\,du \qquad E[X]=\int_a^b x f_X(x)\,dx \qquad \text{median: } F_X(m)=0.5$$

**Kontinuert = INTEGRAL, aldrig sum.**

```python
# === IMPORTS ===
import sympy as sp

# === INDSÆT DINE VÆRDIER HER ===
x = sp.symbols('x', positive=True)
k = sp.symbols('k', positive=True)

basis_udtryk = x            # det x-afhængige udtryk (uden k) i f_X(x) = k * basis_udtryk
lo, hi = 0, 4                # støtte-intervallet [lo, hi]

# === BEREGNING ===
k_loesning = sp.solve(sp.Eq(sp.integrate(k * basis_udtryk, (x, lo, hi)), 1), k)[0]
f = k_loesning * basis_udtryk

x_var = sp.symbols('xvar', positive=True)
F_udtryk = sp.integrate(f.subs(x, x_var), (x_var, lo, x))   # CDF som funktion af x

E_X   = sp.integrate(x * f, (x, lo, hi))
E_X2  = sp.integrate(x**2 * f, (x, lo, hi))
Var_X = sp.simplify(E_X2 - E_X**2)

m = sp.symbols('m', positive=True)
median_kandidater = sp.solve(sp.Eq(F_udtryk.subs(x, m), sp.Rational(1, 2)), m)
median_loesning = median_kandidater[0] if median_kandidater else None

# === RESULTATER ===
print("k        =", k_loesning)
print("f_X(x)   =", f, f"  for {lo} <= x <= {hi}")
print("F_X(x)   =", sp.simplify(F_udtryk))
print("E[X]     =", E_X, "=", float(E_X))
print("Var(X)   =", Var_X, "=", float(Var_X))
print("Median   =", median_loesning, "=", float(median_loesning))
```

### 2.5 Kontinuert CDF givet direkte → find PDF

**Bruges når:** Opgaven giver $F_X(x)$ direkte (i stedet for $f_X(x)$), og du skal finde tæthedsfunktionen, eller en sandsynlighed $P(a<X\le b)$.

$$f_X(x)=F_X'(x) \qquad\qquad P(a<X\le b)=F_X(b)-F_X(a)$$

```python
# === IMPORTS ===
import sympy as sp

# === INDSÆT DINE VÆRDIER HER ===
x = sp.symbols('x')
F_udtryk = (x - 1)**2 / 4    # det indre udtryk for F_X(x) (kun gyldigt i støtte-intervallet)
lo, hi = 1, 3                 # støtte-intervallet, hvor udtrykket ovenfor gælder

a_forespoergsel = 1.5
b_forespoergsel = 2.5

# === BEREGNING ===
f_udtryk = sp.diff(F_udtryk, x)

def F_num(vaerdi):
    if vaerdi <= lo:
        return 0.0
    if vaerdi >= hi:
        return 1.0
    return float(F_udtryk.subs(x, vaerdi))

P_interval = F_num(b_forespoergsel) - F_num(a_forespoergsel)

# === RESULTATER ===
print(f"f_X(x) = {f_udtryk}   for {lo} <= x <= {hi}")
print(f"F_X({a_forespoergsel}) = {F_num(a_forespoergsel):.4f}")
print(f"F_X({b_forespoergsel}) = {F_num(b_forespoergsel):.4f}")
print(f"P({a_forespoergsel} < X <= {b_forespoergsel}) = {P_interval:.4f}")
```

### 2.6 Transformation: lineær (Y=aX+b)

**Bruges når:** $Y=aX+b$ er en simpel lineær omregning af $X$ (fx enhedsskift, standardisering).

$$E[Y]=aE[X]+b \qquad Var(Y)=a^2Var(X) \qquad f_Y(y)=\frac{1}{|a|}f_X\!\left(\frac{y-b}{a}\right)$$

```python
# === IMPORTS ===
import sympy as sp

# === INDSÆT DINE VÆRDIER HER ===
E_X   = 4        # E[X] (kendt fra opgaven eller udregnet separat, se 2.1/2.4)
Var_X = 4/3       # Var(X)
a_koef = 3         # Y = a*X + b
b_koef = 2

# Valgfrit: hvis du også kender f_X(x) symbolsk og vil have f_Y(y) symbolsk
x, y = sp.symbols('x y')
fX_udtryk = sp.Rational(1, 4)   # f_X(x) — sæt til None hvis du ikke skal bruge dette

# === BEREGNING ===
E_Y = a_koef * E_X + b_koef
Var_Y = a_koef**2 * Var_X

if fX_udtryk is not None:
    fY_udtryk = sp.simplify(fX_udtryk.subs(x, (y - b_koef) / a_koef) / abs(a_koef)) if hasattr(fX_udtryk, "subs") else fX_udtryk / abs(a_koef)
else:
    fY_udtryk = None

# === RESULTATER ===
print(f"E[Y]   = {a_koef}*E[X] + {b_koef} = {E_Y}")
print(f"Var(Y) = {a_koef}^2 * Var(X) = {Var_Y}")
if fY_udtryk is not None:
    print(f"f_Y(y) = {fY_udtryk}")
```

### 2.7 Transformation: CDF-metoden (ikke-lineær)

**Bruges når:** $Y=g(X)$ for en **ikke-lineær, monoton** funktion $g$ (fx $Y=X^2$, $Y=1/X$). Tjek altid FØRST om $g$ er voksende eller aftagende — dette snippet gør det automatisk ud fra fortegnet af $g'(x)$.

$$F_Y(y)=\begin{cases}F_X(g^{-1}(y)) & g \text{ voksende}\\ 1-F_X(g^{-1}(y)) & g \text{ aftagende}\end{cases}$$

```python
# === IMPORTS ===
import sympy as sp

# === INDSÆT DINE VÆRDIER HER ===
x, y = sp.symbols('x y', positive=True)
FX_udtryk = x**3          # F_X(x), gyldig i støtte-intervallet
lo, hi = 0, 1               # støtte-intervallet for X
g_udtryk = x**2            # transformationen Y = g(X)

# === BEREGNING ===
g_mrk = sp.diff(g_udtryk, x)
midtpunkt = (lo + hi) / 2
fortegn = g_mrk.subs(x, midtpunkt)
voksende = fortegn > 0

x_fra_y = sp.solve(sp.Eq(g_udtryk, y), x)
# Vælg den løsning, der ligger i støtte-intervallet (typisk den positive/reelle rod)
g_inv = None
for kandidat in x_fra_y:
    testv = kandidat.subs(y, g_udtryk.subs(x, (lo + hi) / 2))
    if testv.is_real:
        g_inv = kandidat
        break
if g_inv is None:
    g_inv = x_fra_y[0]

if voksende:
    FY_udtryk = sp.simplify(FX_udtryk.subs(x, g_inv))
else:
    FY_udtryk = sp.simplify(1 - FX_udtryk.subs(x, g_inv))

fY_udtryk = sp.simplify(sp.diff(FY_udtryk, y))

# === RESULTATER ===
print(f"g'(x) = {g_mrk}  ->  fortegn ved midtpunkt = {fortegn}  ->  g er {'voksende' if voksende else 'aftagende'}")
print(f"g^-1(y) = {g_inv}")
print(f"F_Y(y)  = {FY_udtryk}")
print(f"f_Y(y)  = {fY_udtryk}")
```

### 2.8 Joint kontinuert PDF: c, marginaler, uafhængighed, kovarians

**Bruges når:** To kontinuerte variable har en fælles (joint) tæthedsfunktion $f_{X,Y}(x,y)=c\cdot(\text{udtryk})$ på et rektangel.

$$\iint f_{X,Y}\,dx\,dy=1 \qquad f_X(x)=\int f_{X,Y}\,dy \qquad X\perp Y \iff f_{X,Y}=f_X f_Y \qquad Cov(X,Y)=E[XY]-E[X]E[Y]$$

```python
# === IMPORTS ===
import sympy as sp

# === INDSÆT DINE VÆRDIER HER ===
x, y, c = sp.symbols('x y c', positive=True)
basis_udtryk = x * (1 + y)     # f_{X,Y}(x,y) = c * basis_udtryk
x_lo, x_hi = 0, 2
y_lo, y_hi = 0, 1

# === BEREGNING ===
c_loesning = sp.solve(sp.Eq(sp.integrate(sp.integrate(c * basis_udtryk, (y, y_lo, y_hi)), (x, x_lo, x_hi)), 1), c)[0]
f_joint = c_loesning * basis_udtryk

fX = sp.simplify(sp.integrate(f_joint, (y, y_lo, y_hi)))     # marginal for X
fY = sp.simplify(sp.integrate(f_joint, (x, x_lo, x_hi)))     # marginal for Y

uafhaengige = sp.simplify(fX * fY - f_joint) == 0

E_X  = sp.integrate(x * fX, (x, x_lo, x_hi))
E_Y  = sp.integrate(y * fY, (y, y_lo, y_hi))
E_XY = sp.integrate(sp.integrate(x * y * f_joint, (y, y_lo, y_hi)), (x, x_lo, x_hi))
Cov  = sp.simplify(E_XY - E_X * E_Y)

E_X2 = sp.integrate(x**2 * fX, (x, x_lo, x_hi))
E_Y2 = sp.integrate(y**2 * fY, (y, y_lo, y_hi))
Var_X = sp.simplify(E_X2 - E_X**2)
Var_Y = sp.simplify(E_Y2 - E_Y**2)

# === RESULTATER ===
print("c            =", c_loesning)
print("f_X,Y(x,y)   =", f_joint)
print("f_X(x)       =", fX)
print("f_Y(y)       =", fY)
print("Uafhængige?  =", uafhaengige, " (f_X(x)*f_Y(y) ==? f_X,Y(x,y))")
print("E[X], E[Y]   =", E_X, ",", E_Y)
print("Var(X), Var(Y) =", Var_X, ",", Var_Y)
print("Cov(X,Y)     =", Cov)
```

### 2.9 Joint diskret PMF-tabel: marginaler, betinget fordeling

**Bruges når:** En simultan PMF er givet som en **tabel** for to diskrete variable, evt. med ét manglende felt.

$$f_X(x)=\sum_y f_{X,Y}(x,y) \qquad f_{Y\mid X}(y\mid x)=\frac{f_{X,Y}(x,y)}{f_X(x)} \qquad X\perp Y \iff f_{X,Y}(x,y)=f_X(x)f_Y(y)\ \forall x,y$$

```python
# === IMPORTS ===
# (ingen eksterne pakker nødvendige)

# === INDSÆT DINE VÆRDIER HER ===
X_vals = [-1, 0, 1]
Y_vals = [1, 2, 3, 4]

# Tabellen: (x, y) -> sandsynlighed. Brug None for det manglende felt (hvis der er ét).
tabel = {
    (-1, 1): 1/20, (-1, 2): 3/20, (-1, 3): 2/20, (-1, 4): 0,
    ( 0, 1): 0,    ( 0, 2): 2/20, ( 0, 3): None, ( 0, 4): 1/20,
    ( 1, 1): 4/20, ( 1, 2): 1/20, ( 1, 3): 1/20, ( 1, 4): 2/20,
}

x_betinget = 1   # find f(y | X = x_betinget)

# === BEREGNING ===
kendt_sum = sum(v for v in tabel.values() if v is not None)
manglende_felter = [k for k, v in tabel.items() if v is None]
if manglende_felter:
    manglende_vaerdi = 1 - kendt_sum
    tabel[manglende_felter[0]] = manglende_vaerdi

fX = {xv: sum(tabel[(xv, yv)] for yv in Y_vals) for xv in X_vals}
fY = {yv: sum(tabel[(xv, yv)] for xv in X_vals) for yv in Y_vals}

uafhaengige = all(
    abs(tabel[(xv, yv)] - fX[xv] * fY[yv]) < 1e-9
    for xv in X_vals for yv in Y_vals
)

f_betinget = {yv: tabel[(x_betinget, yv)] / fX[x_betinget] for yv in Y_vals}
E_Y_givet_X = sum(yv * p for yv, p in f_betinget.items())

# === RESULTATER ===
if manglende_felter:
    print(f"Manglende felt {manglende_felter[0]} = {manglende_vaerdi:.4f}")
print("f_X(x):", {k: round(v, 4) for k, v in fX.items()})
print("f_Y(y):", {k: round(v, 4) for k, v in fY.items()})
print("Uafhængige?", uafhaengige)
print(f"f(y | X={x_betinget}):", {k: round(v, 4) for k, v in f_betinget.items()})
print(f"E[Y | X={x_betinget}] = {E_Y_givet_X:.4f}")
```

```python
# === IMPORTS ===
import sympy as sp

# === INDSÆT DINE VÆRDIER HER ===
x, y = sp.symbols('x y', positive=True)
f_joint = x + y     # den (allerede fundne) joint PDF f_{X,Y}(x,y), MED konstanten indsat

# Rektanglet du vil finde sandsynligheden for
a_x, b_x = 0, 0.5
a_y, b_y = 0, 0.5

# === BEREGNING ===
P_rektangel = sp.integrate(sp.integrate(f_joint, (y, a_y, b_y)), (x, a_x, b_x))

# === RESULTATER ===
print(f"P({a_x} < X <= {b_x}, {a_y} < Y <= {b_y}) = {P_rektangel} = {float(P_rektangel):.4f}")
```

### 2.11 Poissonfordeling

**Bruges når:** Antal (sjældne) hændelser i et **fast interval** med kendt gennemsnitsrate — signalord "i gennemsnit ... pr. ...". Husk at **skalere** $\lambda$ til det ønskede intervals længde.

$$P(X=k)=\frac{e^{-\lambda}\lambda^k}{k!}, \qquad E[X]=Var(X)=\lambda$$

```python
# === IMPORTS ===
from scipy.stats import poisson

# === INDSÆT DINE VÆRDIER HER ===
rate = 4              # gennemsnitlig rate, som opgivet (fx pr. minut)
rate_interval = 1.0    # det interval raten er opgivet for (i samme enhed som nedenfor)
oensket_interval = 0.5 # det interval, DU skal bruge (skalerer lambda automatisk)
k = 3                   # antal hændelser du spørger om

# === BEREGNING ===
lam = rate * (oensket_interval / rate_interval)

P_eksakt   = poisson.pmf(k, lam)
P_hoejst_k = poisson.cdf(k, lam)
P_mindst_k = 1 - poisson.cdf(k - 1, lam)

# === RESULTATER ===
print(f"Skaleret lambda for intervallet {oensket_interval} = {lam}")
print(f"P(X = {k})  = {P_eksakt:.4f}")
print(f"P(X <= {k}) = {P_hoejst_k:.4f}")
print(f"P(X >= {k}) = {P_mindst_k:.4f}")
```

### 2.12 Varians af lineær kombination Var(aX+bY)

**Bruges når:** Du skal finde variansen af en sum/lineær kombination af **to** stokastiske variable — husk kovarians-leddet, hvis de ikke er uafhængige.

$$Var(aX+bY)=a^2Var(X)+b^2Var(Y)+2ab\,Cov(X,Y)$$

Hvis $X\perp Y$, er $Cov(X,Y)=0$, og formlen reduceres til det simple additive tilfælde.

```python
# === IMPORTS ===
# (ingen eksterne pakker nødvendige)

# === INDSÆT DINE VÆRDIER HER ===
Var_X = 2/9
Var_Y = 0.5
Cov_XY = 0     # sæt til 0 hvis X og Y er uafhængige
a_koef = 2
b_koef = 1

# === BEREGNING ===
Var_sum = a_koef**2 * Var_X + b_koef**2 * Var_Y + 2 * a_koef * b_koef * Cov_XY

# === RESULTATER ===
print(f"Var({a_koef}X + {b_koef}Y) = {a_koef}^2*Var(X) + {b_koef}^2*Var(Y) + 2*{a_koef}*{b_koef}*Cov(X,Y)")
print(f"                = {a_koef**2}*{Var_X} + {b_koef**2}*{Var_Y} + {2*a_koef*b_koef}*{Cov_XY}")
print(f"                = {Var_sum:.4f}")
```

### 3.1 Konfidensinterval — kendt varians (z)

**Bruges når:** $\sigma^2$ er opgivet/kendt (fx fra en specifikation), og data er normalfordelte eller $n$ er stor nok til CLT.

$$\bar X \pm z_{\alpha/2}\frac{\sigma}{\sqrt n}$$

```python
# === IMPORTS ===
import numpy as np
from scipy.stats import norm

# === INDSÆT DINE VÆRDIER HER ===
xbar = 2955     # stikprøvegennemsnit
sigma = 100      # KENDT populations-standardafvigelse
n = 12            # stikprøvestørrelse
konfidensgrad = 0.95

# === BEREGNING ===
alpha = 1 - konfidensgrad
z_krit = norm.ppf(1 - alpha / 2)
margin = z_krit * sigma / np.sqrt(n)
nedre, ovre = xbar - margin, xbar + margin

# === RESULTATER ===
print(f"z_(alpha/2) = {z_krit:.4f}")
print(f"{int(konfidensgrad*100)}% konfidensinterval: [{nedre:.4f} ; {ovre:.4f}]")
print(f"\nTolkning: Vi er {int(konfidensgrad*100)}% sikre på, at den sande middelværdi ligger mellem {nedre:.4f} og {ovre:.4f}.")
```

### 3.2 Konfidensinterval — ukendt varians (t)

**Bruges når:** $\sigma^2$ er ukendt og skal estimeres fra data ($S$), og $n$ er lille (typisk $<30$) — brug $t$-fordelingen med $n-1$ frihedsgrader.

$$\bar X \pm t_{\alpha/2,\,n-1}\frac{S}{\sqrt n}$$

```python
# === IMPORTS ===
import numpy as np
from scipy.stats import t as t_dist

# === INDSÆT DINE VÆRDIER HER ===
# Mulighed A: du har rådata
data = None   # fx np.array([23.1, 21.4, 22.8, 24.0, 20.9])

# Mulighed B: du har allerede xbar, s, n opsummeret i opgaven (bruges hvis data er None)
xbar_manuel, s_manuel, n_manuel = 497.2, 6.5, 15

konfidensgrad = 0.95

# === BEREGNING ===
if data is not None:
    xbar, s, n = np.mean(data), np.std(data, ddof=1), len(data)
else:
    xbar, s, n = xbar_manuel, s_manuel, n_manuel

alpha = 1 - konfidensgrad
df = n - 1
t_krit = t_dist.ppf(1 - alpha / 2, df)
margin = t_krit * s / np.sqrt(n)
nedre, ovre = xbar - margin, xbar + margin

# === RESULTATER ===
print(f"x̄ = {xbar:.4f}, s = {s:.4f}, n = {n}, df = {df}")
print(f"t_(alpha/2, df) = {t_krit:.4f}")
print(f"{int(konfidensgrad*100)}% konfidensinterval: [{nedre:.4f} ; {ovre:.4f}]")
print(f"\nTolkning: Vi er {int(konfidensgrad*100)}% sikre på, at den sande middelværdi ligger mellem {nedre:.4f} og {ovre:.4f}.")
```

### 3.3 Hypotesetest, én stikprøve — z (kendt varians)

**Bruges når:** Du tester om middelværdien er lig $\mu_0$, og $\sigma^2$ er **kendt**.

$$Z=\frac{\bar X-\mu_0}{\sigma/\sqrt n}$$

**Ensidet vs. tosidet:** sæt `haletype` til `"tosidet"` (er der forskel), `"stoerre"` (er den større end) eller `"mindre"` (er den mindre end).

```python
# === IMPORTS ===
import numpy as np
from scipy.stats import norm

# === INDSÆT DINE VÆRDIER HER ===
xbar = 0.72
mu0 = 0.7
sigma = np.sqrt(0.61)
n = 50
alpha = 0.05
haletype = "tosidet"   # "tosidet", "stoerre" eller "mindre"

# === BEREGNING ===
z = (xbar - mu0) / (sigma / np.sqrt(n))

if haletype == "tosidet":
    p_vaerdi = 2 * (1 - norm.cdf(abs(z)))
    z_krit = norm.ppf(1 - alpha / 2)
    forkast = abs(z) > z_krit
elif haletype == "stoerre":
    p_vaerdi = 1 - norm.cdf(z)
    z_krit = norm.ppf(1 - alpha)
    forkast = z > z_krit
else:  # "mindre"
    p_vaerdi = norm.cdf(z)
    z_krit = norm.ppf(1 - alpha)
    forkast = z < -z_krit

# === RESULTATER ===
print(f"z-teststørrelse = {z:.4f}")
print(f"kritisk værdi   = {z_krit:.4f}  ({haletype})")
print(f"p-værdi         = {p_vaerdi:.4f}")
print()
if forkast:
    print(f"p-værdi < {alpha}  ->  FORKAST H0 på {int(alpha*100)}% signifikansniveau.")
else:
    print(f"p-værdi >= {alpha}  ->  KAN IKKE forkaste H0 på {int(alpha*100)}% signifikansniveau.")
```

<a id="3.4"></a>

### 3.4 Hypotesetest, én stikprøve — t (ukendt varians)

**Bruges når:** Du tester om middelværdien er lig $\mu_0$, og $\sigma^2$ er **ukendt** (skal estimeres fra data).

$$T=\frac{\bar X-\mu_0}{S/\sqrt n}\sim t_{n-1}$$

```python
# === IMPORTS ===
import numpy as np
from scipy.stats import t as t_dist

# === INDSÆT DINE VÆRDIER HER ===
# Mulighed A: rådata
data = None   # fx np.array([14.2, 15.8, 13.5, 16.1])

# Mulighed B: opsummerede tal (bruges hvis data er None)
xbar_manuel, s_manuel, n_manuel = 497.2, 6.5, 15

mu0 = 500
alpha = 0.05
haletype = "tosidet"   # "tosidet", "stoerre" eller "mindre"

# === BEREGNING ===
if data is not None:
    xbar, s, n = np.mean(data), np.std(data, ddof=1), len(data)
else:
    xbar, s, n = xbar_manuel, s_manuel, n_manuel

df = n - 1
t_stat = (xbar - mu0) / (s / np.sqrt(n))

if haletype == "tosidet":
    p_vaerdi = 2 * (1 - t_dist.cdf(abs(t_stat), df))
    t_krit = t_dist.ppf(1 - alpha / 2, df)
    forkast = abs(t_stat) > t_krit
elif haletype == "stoerre":
    p_vaerdi = 1 - t_dist.cdf(t_stat, df)
    t_krit = t_dist.ppf(1 - alpha, df)
    forkast = t_stat > t_krit
else:  # "mindre"
    p_vaerdi = t_dist.cdf(t_stat, df)
    t_krit = t_dist.ppf(1 - alpha, df)
    forkast = t_stat < -t_krit

# === RESULTATER ===
print(f"x̄ = {xbar:.4f}, s = {s:.4f}, n = {n}, df = {df}")
print(f"t-teststørrelse = {t_stat:.4f}")
print(f"kritisk værdi   = {t_krit:.4f}  ({haletype})")
print(f"p-værdi         = {p_vaerdi:.4f}")
print()
if forkast:
    print(f"|t| > t*  ->  FORKAST H0 på {int(alpha*100)}% signifikansniveau.")
else:
    print(f"|t| <= t*  ->  KAN IKKE forkaste H0 på {int(alpha*100)}% signifikansniveau.")
```

### 3.5 To uafhængige stikprøver — pooled varians

**Bruges når:** To **uafhængige, uparrede** stikprøver, ukendt men **antaget ens** varians i begge populationer. Giver både hypotesetest og konfidensinterval for forskellen.

$$S_p^2=\frac{(n_1-1)S_1^2+(n_2-1)S_2^2}{n_1+n_2-2} \qquad T=\frac{(\bar X_1-\bar X_2)-\delta_0}{S_p\sqrt{\tfrac1{n_1}+\tfrac1{n_2}}}\sim t_{n_1+n_2-2}$$

```python
# === IMPORTS ===
import numpy as np
from scipy.stats import t as t_dist

# === INDSÆT DINE VÆRDIER HER ===
# Mulighed A: rådata for de to grupper
data1 = None   # fx np.array([...])
data2 = None

# Mulighed B: opsummerede tal (bruges hvis data1/data2 er None)
n1_manuel, xbar1_manuel, s1_2_manuel = 14, 120.5, 18.2   # s1_2 = s1^2 (varians, IKKE std)
n2_manuel, xbar2_manuel, s2_2_manuel = 11, 115.8, 14.7

delta0 = 0            # H0: mu1 - mu2 = delta0 (normalt 0)
alpha = 0.05
haletype = "tosidet"   # "tosidet", "stoerre" (mu1>mu2) eller "mindre" (mu1<mu2)
konfidensgrad = 0.95

# === BEREGNING ===
if data1 is not None and data2 is not None:
    n1, xbar1, s1_2 = len(data1), np.mean(data1), np.var(data1, ddof=1)
    n2, xbar2, s2_2 = len(data2), np.mean(data2), np.var(data2, ddof=1)
else:
    n1, xbar1, s1_2 = n1_manuel, xbar1_manuel, s1_2_manuel
    n2, xbar2, s2_2 = n2_manuel, xbar2_manuel, s2_2_manuel

df = n1 + n2 - 2
Sp2 = ((n1 - 1) * s1_2 + (n2 - 1) * s2_2) / df
Sp = np.sqrt(Sp2)
SE = Sp * np.sqrt(1/n1 + 1/n2)
t_stat = ((xbar1 - xbar2) - delta0) / SE

if haletype == "tosidet":
    p_vaerdi = 2 * (1 - t_dist.cdf(abs(t_stat), df))
    t_krit_test = t_dist.ppf(1 - alpha / 2, df)
    forkast = abs(t_stat) > t_krit_test
elif haletype == "stoerre":
    p_vaerdi = 1 - t_dist.cdf(t_stat, df)
    t_krit_test = t_dist.ppf(1 - alpha, df)
    forkast = t_stat > t_krit_test
else:
    p_vaerdi = t_dist.cdf(t_stat, df)
    t_krit_test = t_dist.ppf(1 - alpha, df)
    forkast = t_stat < -t_krit_test

alpha_ci = 1 - konfidensgrad
t_krit_ci = t_dist.ppf(1 - alpha_ci / 2, df)   # KI er altid tosidet, uanset test
margin = t_krit_ci * SE
nedre, ovre = (xbar1 - xbar2) - margin, (xbar1 - xbar2) + margin

# === RESULTATER ===
print(f"Sp^2 = {Sp2:.4f}, df = {df}")
print(f"t-teststørrelse = {t_stat:.4f}, kritisk værdi = {t_krit_test:.4f} ({haletype})")
print(f"p-værdi = {p_vaerdi:.4f}")
print("FORKAST H0" if forkast else "KAN IKKE forkaste H0", f"på {int(alpha*100)}% niveau.")
print()
print(f"{int(konfidensgrad*100)}% konfidensinterval for (mu1-mu2): [{nedre:.4f} ; {ovre:.4f}]")
```

### 3.6 Parrede stikprøver

**Bruges når:** Data er **parrede** (samme forsøgsenhed/lokation målt to gange, fx før/efter). Reducér altid til differencer $D_i$ og brug en ét-stikprøve $t$-test på dem.

$$D_i=X_{1i}-X_{2i} \qquad T=\frac{\bar D-\delta_0}{S_D/\sqrt n}\sim t_{n-1}$$

```python
# === IMPORTS ===
import numpy as np
from scipy.stats import t as t_dist

# === INDSÆT DINE VÆRDIER HER ===
x1 = np.array([412, 398, 405, 420, 390, 415, 408, 402, 411, 399])   # fx "ny mølle"
x2 = np.array([405, 390, 400, 410, 385, 408, 400, 395, 402, 392])   # fx "gammel mølle"

delta0 = 0
alpha = 0.05
haletype = "tosidet"
konfidensgrad = 0.95

# === BEREGNING ===
D = x1 - x2
n = len(D)
Dbar = np.mean(D)
SD = np.std(D, ddof=1)
df = n - 1

t_stat = (Dbar - delta0) / (SD / np.sqrt(n))

if haletype == "tosidet":
    p_vaerdi = 2 * (1 - t_dist.cdf(abs(t_stat), df))
    t_krit_test = t_dist.ppf(1 - alpha / 2, df)
    forkast = abs(t_stat) > t_krit_test
elif haletype == "stoerre":
    p_vaerdi = 1 - t_dist.cdf(t_stat, df)
    t_krit_test = t_dist.ppf(1 - alpha, df)
    forkast = t_stat > t_krit_test
else:
    p_vaerdi = t_dist.cdf(t_stat, df)
    t_krit_test = t_dist.ppf(1 - alpha, df)
    forkast = t_stat < -t_krit_test

alpha_ci = 1 - konfidensgrad
t_krit_ci = t_dist.ppf(1 - alpha_ci / 2, df)
margin = t_krit_ci * SD / np.sqrt(n)
nedre, ovre = Dbar - margin, Dbar + margin

# === RESULTATER ===
print(f"D̄ = {Dbar:.4f}, S_D = {SD:.4f}, n = {n}, df = {df}")
print(f"t-teststørrelse = {t_stat:.4f}, kritisk værdi = {t_krit_test:.4f} ({haletype})")
print(f"p-værdi = {p_vaerdi:.4f}")
print("FORKAST H0" if forkast else "KAN IKKE forkaste H0", f"på {int(alpha*100)}% niveau.")
print()
print(f"{int(konfidensgrad*100)}% konfidensinterval for middel-differencen: [{nedre:.4f} ; {ovre:.4f}]")
```

### 3.7 Lineær regression

**Bruges når:** Du skal estimere en lineær sammenhæng $y_i=\beta_0+\beta_1x_i+\varepsilon_i$ mellem to variable ud fra datapunkter, teste om hældningen er signifikant, og/eller angive et konfidensinterval for $\beta_1$.

$$\hat\beta_1=\frac{\sum(x_i-\bar x)(y_i-\bar y)}{\sum(x_i-\bar x)^2} \qquad \hat\beta_0=\bar y-\hat\beta_1\bar x \qquad T=\frac{\hat\beta_1-0}{SE(\hat\beta_1)}\sim t_{n-2}$$

```python
# === IMPORTS ===
import numpy as np
from scipy.stats import t as t_dist

# === INDSÆT DINE VÆRDIER HER ===
x = np.array([20, 25, 30, 35, 40])
y = np.array([3.1, 4.0, 5.2, 6.8, 8.5])
alpha = 0.05
konfidensgrad = 0.95

# === BEREGNING ===
n = len(x)
xbar, ybar = np.mean(x), np.mean(y)

Sxy = np.sum((x - xbar) * (y - ybar))
Sxx = np.sum((x - xbar) ** 2)

beta1 = Sxy / Sxx
beta0 = ybar - beta1 * xbar

y_hat = beta0 + beta1 * x
residualer = y - y_hat
SSE = np.sum(residualer ** 2)
SST = np.sum((y - ybar) ** 2)
R2 = 1 - SSE / SST

df = n - 2
s2 = SSE / df
SE_beta1 = np.sqrt(s2 / Sxx)

t_stat = beta1 / SE_beta1
p_vaerdi = 2 * (1 - t_dist.cdf(abs(t_stat), df))
t_krit_test = t_dist.ppf(1 - alpha / 2, df)

alpha_ci = 1 - konfidensgrad
t_krit_ci = t_dist.ppf(1 - alpha_ci / 2, df)
margin = t_krit_ci * SE_beta1
nedre, ovre = beta1 - margin, beta1 + margin

# === RESULTATER ===
print(f"beta0 = {beta0:.4f}")
print(f"beta1 = {beta1:.4f}")
print(f"Fittet linje: y_hat = {beta0:.4f} + {beta1:.4f}*x")
print(f"Residualer: {np.round(residualer, 4)}  (sanity-check: sum ≈ 0 -> {residualer.sum():.2e})")
print(f"R^2 = {R2:.4f}")
print()
print(f"SE(beta1) = {SE_beta1:.5f}")
print(f"t-teststørrelse (H0: beta1=0) = {t_stat:.4f}, kritisk værdi = {t_krit_test:.4f}")
print(f"p-værdi = {p_vaerdi:.6f}")
print("FORKAST H0 (signifikant sammenhæng)" if abs(t_stat) > t_krit_test else "KAN IKKE forkaste H0")
print()
print(f"{int(konfidensgrad*100)}% konfidensinterval for beta1: [{nedre:.4f} ; {ovre:.4f}]")
```

### 3.8 Punktestimator, bias og simulering

**Bruges når:** Opgaven beder dig opstille en punktestimator (fx $\hat p=Y/n$), vise om den er unbiased, og/eller **verificere** dette ved simulering (store tals lov).

```python
# === IMPORTS ===
import numpy as np

# === INDSÆT DINE VÆRDIER HER ===
p_sand = 0.018     # den "sande" parameterværdi, du simulerer ud fra
n = 3000            # antal enheder pr. stikprøve
N = 10000            # antal gentagne stikprøver (simuleringer)
seed = 42

# === BEREGNING ===
rng = np.random.default_rng(seed)
Y = rng.binomial(n, p_sand, size=N)   # antal succeser i hver af de N stikprøver
p_hat = Y / n                          # punktestimator for hver stikprøve

gennemsnit_phat = np.mean(p_hat)
bias_simuleret = gennemsnit_phat - p_sand

# === RESULTATER ===
print(f"Sand værdi p           = {p_sand}")
print(f"Gennemsnit af p_hat    = {gennemsnit_phat:.5f}  (over {N} simuleringer)")
print(f"Simuleret bias         = {bias_simuleret:.5f}   (bør ligge tæt på 0, da E[p_hat]=p er unbiased)")
print()
print("Teoretisk: E[p_hat] = E[Y]/n = np/n = p  ->  Bias(p_hat) = 0 (unbiased)")
```

### 3.9 Invers transformationsmetode + CI af simulerede data

**Bruges når:** Du skal **simulere** stikprøver fra en kendt kontinuert fordeling ud fra uniforme tilfældstal, og evt. bygge et konfidensinterval ud fra de simulerede data.

$$U\sim\mathrm{Uniform}(0,1), \qquad X=F_X^{-1}(U) \implies X\sim f_X$$

```python
# === IMPORTS ===
import numpy as np
import sympy as sp
from scipy.stats import norm

# === INDSÆT DINE VÆRDIER HER ===
x, u = sp.symbols('x u', positive=True)
fX_udtryk = 3 * x**2   # tæthedsfunktionen f_X(x)
lo, hi = 0, 1            # støtte-intervallet

N = 100        # antal simulerede værdier
seed = 123
konfidensgrad = 0.95

# === BEREGNING ===
xh = sp.symbols('xh', positive=True)
F_udtryk = sp.integrate(fX_udtryk.subs(x, xh), (xh, lo, x))   # F_X(x)

kandidater = sp.solve(sp.Eq(F_udtryk, u), x)
F_inv_udtryk = None
for kandidat in kandidater:
    testv = complex(kandidat.subs(u, 0.5))
    if abs(testv.imag) < 1e-9 and lo <= testv.real <= hi:
        F_inv_udtryk = kandidat
        break
if F_inv_udtryk is None:
    F_inv_udtryk = kandidater[0]

F_inv_func = sp.lambdify(u, F_inv_udtryk, "numpy")

rng = np.random.default_rng(seed)
U = rng.uniform(0, 1, N)
X_sim = F_inv_func(U)

xbar = np.mean(X_sim)
s = np.std(X_sim, ddof=1)
alpha = 1 - konfidensgrad
z_krit = norm.ppf(1 - alpha / 2)   # brug z siden n er "stort" (>=30); brug t_dist hvis lille n
margin = z_krit * s / np.sqrt(N)
nedre, ovre = xbar - margin, xbar + margin

# === RESULTATER ===
print(f"F_X(x)     = {sp.simplify(F_udtryk)}")
print(f"F_X^-1(u)  = {F_inv_udtryk}")
print()
print(f"Simuleret x̄ = {xbar:.4f}, s = {s:.4f}  (N={N})")
print(f"{int(konfidensgrad*100)}% konfidensinterval: [{nedre:.4f} ; {ovre:.4f}]")
print()
print("Tolkning: sammenlign intervallet med den teoretiske E[X] (se 2.4) — ligger den indenfor, er simuleringen konsistent med modellen.")
```

### 3.10 Box–Muller-transformation

**Bruges når:** Du skal simulere **standardnormalfordelte** tal, hvor invers transformation ikke virker direkte (fordi $\Phi^{-1}$ ikke har et lukket udtryk).

$$Z_0=\sqrt{-2\ln U_1}\cos(2\pi U_2), \qquad Z_1=\sqrt{-2\ln U_1}\sin(2\pi U_2)$$

```python
# === IMPORTS ===
import numpy as np

# === INDSÆT DINE VÆRDIER HER ===
U1, U2 = 0.25, 0.75   # dine uniforme tal (mellem 0 og 1) -- kan også være arrays til flere på én gang
mu, sigma = 3000, 100   # ønsket N(mu, sigma^2), fx en batterilevetid

# === BEREGNING ===
Z0 = np.sqrt(-2 * np.log(U1)) * np.cos(2 * np.pi * U2)
Z1 = np.sqrt(-2 * np.log(U1)) * np.sin(2 * np.pi * U2)

T0 = mu + sigma * Z0
T1 = mu + sigma * Z1

# === RESULTATER ===
print(f"Z0 = {Z0}")
print(f"Z1 = {Z1}")
print(f"T (fra Z0) = mu + sigma*Z0 = {T0}")
print(f"T (fra Z1) = mu + sigma*Z1 = {T1}")
```

### 3.11 Kombineret: diskret PMF → simulation → test → interval

**Bruges når:** En opgave kæder flere metoder sammen i forlængelse af hinanden — typisk: opstil en diskret PMF/model → simulér data fra den → test om den simulerede middelværdi er konsistent med modellen → byg et konfidensinterval.

**Vigtigt:** Da $\sigma^2$ her kommer fra **modellen** (ikke estimeres fra data via $S^2$), bruges **z**, ikke t — uanset hvor lille $n$ er. Se Eksamensbogen Del IV, Trin 5.

```python
# === IMPORTS ===
import numpy as np
from scipy.stats import norm

# === INDSÆT DINE VÆRDIER HER ===
xs = [0, 1, 2]
ps = [0.5, 0.3, 0.2]   # den diskrete model-PMF

N = 50          # antal simulerede værdier
seed = 42
alpha = 0.05
konfidensgrad = 0.95

# === BEREGNING ===
# 1) Teoretisk middelværdi og varians fra modellen (diskret -> sum, se 2.1)
mu0   = sum(x * p for x, p in zip(xs, ps))
E_X2  = sum(x**2 * p for x, p in zip(xs, ps))
sigma2 = E_X2 - mu0**2
sigma = np.sqrt(sigma2)

# 2) Simulér N værdier med invers CDF-metoden (diskret, se 2.2)
cdf = np.cumsum(ps)
rng = np.random.default_rng(seed)
U = rng.uniform(0, 1, N)
X_sim = np.array([xs[np.searchsorted(cdf, u)] for u in U])

xbar_sim = np.mean(X_sim)

# 3) Hypotesetest: H0: mu = mu0, med KENDT sigma (fra modellen) -> z-test
z_stat = (xbar_sim - mu0) / (sigma / np.sqrt(N))
p_vaerdi = 2 * (1 - norm.cdf(abs(z_stat)))
z_krit = norm.ppf(1 - alpha / 2)
forkast = abs(z_stat) > z_krit

# 4) Konfidensinterval, samme kendte sigma
alpha_ci = 1 - konfidensgrad
z_krit_ci = norm.ppf(1 - alpha_ci / 2)
margin = z_krit_ci * sigma / np.sqrt(N)
nedre, ovre = xbar_sim - margin, xbar_sim + margin

# === RESULTATER ===
print(f"Model:  mu0 = {mu0:.4f},  sigma^2 = {sigma2:.4f},  sigma = {sigma:.4f}")
print(f"Simuleret x̄ (N={N}) = {xbar_sim:.4f}")
print()
print(f"z-teststørrelse = {z_stat:.4f}, kritisk værdi = {z_krit:.4f}")
print(f"p-værdi = {p_vaerdi:.4f}")
print("FORKAST H0" if forkast else "KAN IKKE forkaste H0", f"-> simuleringen {'strider imod' if forkast else 'er konsistent med'} modellen.")
print()
print(f"{int(konfidensgrad*100)}% konfidensinterval for middelværdien: [{nedre:.4f} ; {ovre:.4f}]")
print(f"(Sanity-check: indeholder intervallet mu0={mu0:.4f}? -> {nedre <= mu0 <= ovre}  — bør stemme overens med test-konklusionen)")
```

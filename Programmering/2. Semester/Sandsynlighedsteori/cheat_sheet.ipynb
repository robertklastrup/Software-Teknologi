# 1. Sandsynlighedsteori

## 1.1 Unions-formlen

Bruges til at finde sandsynligheden for at **A eller B** sker.

**Indsæt:**
- `P_A` = sandsynlighed for hændelse A
- `P_B` = sandsynlighed for hændelse B
- `P_A_og_B` = sandsynlighed for at begge sker samtidig

$$P(A \cup B) = P(A) + P(B) - P(A \cap B)$$

```python
# ── INDSÆT DINE TAL HER ──────────────────────────
P_A     = 0.08   # P(A) — fx P(bakterier)
P_B     = 0.05   # P(B) — fx P(kemikalier)
P_A_og_B = 0.02  # P(A ∩ B) — fx P(bakterier OG kemikalier)
# ─────────────────────────────────────────────────

P_union = P_A + P_B - P_A_og_B
P_hverken = 1 - P_union

print(f"P(A ∪ B)     = {P_union:.4f}  →  {P_union*100:.2f}%")
print(f"P(hverken)   = {P_hverken:.4f}  →  {P_hverken*100:.2f}%")
```

## 1.2 Total sandsynlighed

Bruges når udfaldsrummet er opdelt i to dele (B og Bᶜ), og vi vil finde den samlede sandsynlighed for A.

**Indsæt:**
- `P_A_givet_B` = P(A | B) — sandsynlighed for A når B er indtruffet
- `P_A_givet_Bc` = P(A | Bᶜ) — sandsynlighed for A når B *ikke* er indtruffet
- `P_B` = sandsynlighed for B (prior)

$$P(A) = P(A \mid B) \cdot P(B) + P(A \mid B^c) \cdot P(B^c)$$

```python
# ── INDSÆT DINE TAL HER ──────────────────────────
P_A_givet_B  = 0.94   # P(A|B)  — fx P(alarm | bakterier)
P_A_givet_Bc = 0.04   # P(A|Bᶜ) — fx P(alarm | ingen bakterier)
P_B          = 0.08   # P(B)    — fx P(bakterier)
# ─────────────────────────────────────────────────

P_Bc = 1 - P_B
P_A  = P_A_givet_B * P_B + P_A_givet_Bc * P_Bc

print(f"P(Bᶜ)  = {P_Bc:.4f}")
print(f"P(A)   = {P_A_givet_B} × {P_B} + {P_A_givet_Bc} × {P_Bc:.4f}")
print(f"P(A)   = {P_A:.4f}  →  {P_A*100:.2f}%")
```

## 1.3 Bayes' formel

Bruges til at "vende" en betinget sandsynlighed — vi kender P(A|B) men vil finde P(B|A).

**Indsæt:**
- `P_A_givet_B` = P(A | B) — den kendte betingede sandsynlighed
- `P_B` = prior sandsynlighed for B
- `P_A` = den samlede P(A), typisk beregnet med total sandsynlighed

$$P(B \mid A) = P(A \mid B) \cdot \frac{P(B)}{P(A)}$$

```python
# ── INDSÆT DINE TAL HER ──────────────────────────
P_A_givet_B = 0.94   # P(A|B)  — fx P(alarm | bakterier)
P_B         = 0.08   # P(B)    — fx P(bakterier)
P_A         = 0.112  # P(A)    — beregnet med total sandsynlighed
# ─────────────────────────────────────────────────

P_B_givet_A = P_A_givet_B * P_B / P_A

print(f"P(B|A) = {P_A_givet_B} × {P_B} / {P_A}")
print(f"P(B|A) = {P_B_givet_A:.4f}  →  {P_B_givet_A*100:.2f}%")
```

## 1.4 Binomialfordeling

Bruges til at finde sandsynlighed for **præcis k succeser** i n uafhængige forsøg med succes-sandsynlighed p.
Brug også til P(X ≤ k) ved at lægge enkelt-sandsynligheder sammen, eller brug `stats.binom.cdf`.

**Indsæt:**
- `n` = antal forsøg
- `k` = ønsket antal succeser
- `p` = succes-sandsynlighed per forsøg

$$P(X = k) = \binom{n}{k} \cdot p^k \cdot (1-p)^{n-k}$$
$$E[X] = n \cdot p, \quad \text{Var}(X) = n \cdot p \cdot (1-p)$$

```python
# ── INDSÆT DINE TAL HER ──────────────────────────
n = 7      # antal forsøg
k = 2      # ønsket antal succeser (for præcis k)
p = 0.112   # succes-sandsynlighed per forsøg
k_max = 1  # øvre grænse for P(X ≤ k_max)
# ─────────────────────────────────────────────────

binom_koeff = comb(n, k)
P_praecis_k = binom_koeff * p**k * (1-p)**(n-k)
P_hoejst_k  = stats.binom.cdf(k_max, n, p)   # P(X ≤ k_max)
E_X         = n * p
Var_X       = n * p * (1-p)

print(f"Binomialkoefficient C({n},{k}) = {binom_koeff}")
print(f"P(X = {k})     = {P_praecis_k:.4f}  →  {P_praecis_k*100:.2f}%")
print(f"P(X ≤ {k_max})    = {P_hoejst_k:.4f}  →  {P_hoejst_k*100:.2f}%")
print(f"E[X]           = {E_X:.4f}")
print(f"Var(X)         = {Var_X:.4f}")
```

## 1.5 Poissonfordeling

Bruges til at tælle begivenheder i et tidsinterval med konstant rate λ.
**Vigtig regel:** Hvis du skifter tidsinterval, skalér λ proportionalt — fx halvt interval → λ/2.

**Indsæt:**
- `lam` = λ (gennemsnitlig rate for det valgte tidsinterval)
- `k` = ønsket antal begivenheder

$$P(X = k) = \frac{e^{-\lambda} \cdot \lambda^k}{k!}$$
$$P(X \geq 1) = 1 - P(X = 0)$$

```python
# ── INDSÆT DINE TAL HER ──────────────────────────
lam    = 3     # λ — gennemsnitlig rate (husk at skalér ved andet tidsinterval)
k      = 2     # præcis k begivenheder
k_min  = 1     # mindst k_min begivenheder  →  P(X ≥ k_min)
# ─────────────────────────────────────────────────

P_praecis = stats.poisson.pmf(k, lam)
P_mindst  = 1 - stats.poisson.cdf(k_min - 1, lam)   # P(X ≥ k_min)
P_ingen   = stats.poisson.pmf(0, lam)                # P(X = 0)

print(f"λ = {lam}")
print(f"P(X = {k})       = {P_praecis:.4f}  →  {P_praecis*100:.2f}%")
print(f"P(X ≥ {k_min})      = {P_mindst:.4f}  →  {P_mindst*100:.2f}%")
print(f"P(X = 0)        = {P_ingen:.4f}  →  {P_ingen*100:.2f}%")
print()
print("--- Skalering af λ til andet tidsinterval ---")
faktor = 0.5   # ← ændr dette: 0.5 = halvt interval, 0.25 = kvarter osv.
lam_ny = lam * faktor
print(f"λ × {faktor} = {lam_ny}  (brug dette ved {faktor} af det originale interval)")
```

## 2.1 Verificering af PDF + CDF

Bruges til at tjekke at en tæthedsfunktion er gyldig (integral = 1), og til at beregne fordelingsfunktionen og sandsynligheder.

**Indsæt:**
- `f` = din tæthedsfunktion som Python-funktion
- `a`, `b` = støtteintervallets nedre og øvre grænse
- `x1`, `x2` = grænser for sandsynlighed P(x1 < X ≤ x2)

$$\int_a^b f_X(x)\,dx = 1 \quad \text{(gyldig PDF)}$$
$$P(x_1 < X \leq x_2) = F_X(x_2) - F_X(x_1) = \int_{x_1}^{x_2} f_X(x)\,dx$$

```python
from scipy.integrate import quad

# ── INDSÆT DINE TAL HER ──────────────────────────
def f(x):
    return 3/32*(4*x-x**2)   # ← ændr til din tæthedsfunktion

a  = 0    # nedre grænse for støtteintervallet
b  = 4    # øvre grænse for støtteintervallet
x1 = 1    # nedre grænse for sandsynlighed
x2 = 3    # øvre grænse for sandsynlighed
# ─────────────────────────────────────────────────

integral, _ = quad(f, a, b)
P_interval, _ = quad(f, x1, x2)

print(f"∫ f(x) dx fra {a} til {b} = {integral:.6f}")
if abs(integral - 1.0) < 1e-6:
    print("✅ f(x) er en gyldig tæthedsfunktion (integral = 1)")
else:
    print(f"❌ Integral = {integral:.4f} — ikke en gyldig PDF!")

print()
print(f"P({x1} < X ≤ {x2}) = {P_interval:.4f}  →  {P_interval*100:.2f}%")
```

## 2.1b Fordelingsfunktion F_X(x) — tabel over alle værdier

Bruges til at beregne CDF'en F_X(x) = P(X ≤ x) for ét bestemt x, og som tabel for hele støtteintervallet.

**Indsæt:**
- `f` = din tæthedsfunktion
- `a`, `b` = støtteintervallets grænser
- `x_punkt` = det specifikke x du vil beregne F_X(x) for

$$F_X(x) = \int_a^x f_X(t)\,dt$$

Udenfor støtteintervallet: F_X(x) = 0 for x < a, og F_X(x) = 1 for x > b.

```python
import sys, subprocess
subprocess.run([sys.executable, "-m", "pip", "install", "sympy", "-q"], check=True)
print("sympy klar ✅")
```

```python
from scipy.integrate import quad
from sympy import *

# ── INDSÆT DINE TAL HER ──────────────────────────
def f(x):
    return (4/27)*x**2 * (3-x)   # ← ændr til din tæthedsfunktion

a       = 0    # nedre grænse for støtteintervallet
b       = 3    # øvre grænse for støtteintervallet
x_punkt = 3    # det x du vil beregne F_X(x) for
# ─────────────────────────────────────────────────

# ── Symbolsk udledning af F_X(x) ──────────────────
x, t = symbols('x t', real=True)

# Udtryk tæthedsfunktionen symbolsk (matcher f ovenfor)
f_sym = Rational(4, 27) * x**2 * (3 - x)

print("═" * 50)
print("SYMBOLSK UDLEDNING AF F_X(x)")
print("═" * 50)
print()
print("Trin 1 — Tæthedsfunktionen:")
print(f"  f_X(t) = {f_sym}")
print()

# Trin 2: Ubestemt integral
F_ubestemt = integrate(f_sym, t)
print("Trin 2 — Ubestemt integral ∫ f_X(t) dt:")
print(f"  ∫ f_X(t) dt = {F_ubestemt}")
print()

# Trin 3: Indsæt grænser (fra a til x)
F_sym = integrate(f_sym, (t, a, x))
F_sym_simplified = simplify(F_sym)
print(f"Trin 3 — F_X(x) = ∫ fra {a} til x:")
print(f"  F_X(x) = [{F_ubestemt}] fra {a} til x")
print(f"  F_X(x) = {F_sym_simplified}")
print()
print(f"Samlet udtryk for {a} ≤ x ≤ {b}:")
print(f"  F_X(x) = {F_sym_simplified}")
print()
print("Stykvis definition:")
print(f"  F_X(x) = 0                     for x < {a}")
print(f"  F_X(x) = {F_sym_simplified}   for {a} ≤ x ≤ {b}")
print(f"  F_X(x) = 1                     for x > {b}")

# ── Numerisk beregning ─────────────────────────────
print()
print("═" * 50)
print("NUMERISK BEREGNING")
print("═" * 50)

def F_X(x_val):
    if x_val <= a:
        return 0.0
    if x_val >= b:
        return 1.0
    val, _ = quad(f, a, x_val)
    return val

print(f"\nF_X({x_punkt}) = P(X ≤ {x_punkt}) = {F_X(x_punkt):.4f}  →  {F_X(x_punkt)*100:.2f}%")
print()

import numpy as np
print(f"{'x':>6}  {'F_X(x)':>8}  {'%':>7}")
print("─" * 28)
for x_val in np.linspace(a, b, 9):
    fx = F_X(x_val)
    print(f"{x_val:>6.3f}  {fx:>8.4f}  {fx*100:>6.2f}%")
```

## 2.2 Middelværdi og varians — kontinuert

Bruges til at beregne E[X] og Var(X) for en kontinuert stokastisk variabel.

**Indsæt:**
- `f` = din tæthedsfunktion
- `a`, `b` = støtteintervallet

$$E[X] = \int_a^b x \cdot f_X(x)\,dx$$
$$E[X^2] = \int_a^b x^2 \cdot f_X(x)\,dx$$
$$\text{Var}(X) = E[X^2] - (E[X])^2$$

```python
from scipy.integrate import quad

# ── INDSÆT DINE TAL HER ──────────────────────────
def f(x):
    return 3/32 * (4*x - x**2)   # ← ændr til din tæthedsfunktion

a = 0   # nedre grænse
b = 4   # øvre grænse
# ─────────────────────────────────────────────────

E_X,   _ = quad(lambda x: x    * f(x), a, b)
E_X2,  _ = quad(lambda x: x**2 * f(x), a, b)
Var_X     = E_X2 - E_X**2

print(f"E[X]   = ∫ x·f(x) dx  = {E_X:.4f}")
print(f"E[X²]  = ∫ x²·f(x) dx = {E_X2:.4f}")
print(f"Var(X) = E[X²] - (E[X])² = {E_X2:.4f} - {E_X**2:.4f} = {Var_X:.4f}")
print(f"Std(X) = √Var(X) = {Var_X**0.5:.4f}")
```

## 2.3 Lineær transformation — Y = aX + b

Bruges til at beregne E[Y] og Var(Y), samt bestemme tæthedsfunktionen f_Y(y) og dens støtteinterval, når Y = aX + b er en lineær funktion af X.

**Indsæt:**
- `a` = skaleringskoefficient
- `b` = konstantled
- `E_X`, `Var_X` = middelværdi og varians af X
- `f_X_sym` = tæthedsfunktionen for X som sympy-udtryk (brug symbolet `x`)
- `x_min`, `x_max` = støtteintervallet for X

$$E[aX + b] = a \cdot E[X] + b$$
$$\text{Var}(aX + b) = a^2 \cdot \text{Var}(X)$$

**Transformation af tæthedsfunktion (change-of-variables):**
$$f_Y(y) = f_X\!\left(\frac{y - b}{a}\right) \cdot \frac{1}{|a|}$$

Støtteintervallet for Y bestemmes ud fra a's fortegn:
$$\text{Støtte for } Y = \begin{cases} [a \cdot x_{\min} + b,\ a \cdot x_{\max} + b] & a > 0 \\ [a \cdot x_{\max} + b,\ a \cdot x_{\min} + b] & a < 0 \end{cases}$$

```python
from sympy import *

# ── INDSÆT DINE TAL HER ──────────────────────────
a     = 3      # skaleringskoefficient i Y = aX + b
b     = 2      # konstantled
E_X   = 4      # middelværdi af X
Var_X = 1    # varians af X

x_min = 2      # nedre grænse for støtteintervallet for X
x_max = 6      # øvre grænse for støtteintervallet for X

x = symbols('x', real=True)
f_X_sym = Rational(1/4)   # ← ændr til din f_X(x) med symbolet x
# ─────────────────────────────────────────────────

print("═" * 55)
print("TRIN 1 — Middelværdi og varians af Y = aX + b")
print("═" * 55)
E_Y   = a * E_X + b
Var_Y = a**2 * Var_X
print(f"Y = {a}X + ({b})")
print(f"E[Y]   = {a} · {E_X} + ({b}) = {E_Y:.4f}")
print(f"Var(Y) = {a}² · {Var_X} = {Var_Y:.4f}")

print()
print("═" * 55)
print("TRIN 2 — Støtteinterval for Y")
print("═" * 55)
if a > 0:
    y_min = a * x_min + b
    y_max = a * x_max + b
    print(f"a = {a} > 0  →  støtten skaleres og forskydes proportionalt")
    print(f"y_min = {a} · {x_min} + ({b}) = {y_min}")
    print(f"y_max = {a} · {x_max} + ({b}) = {y_max}")
else:
    y_min = a * x_max + b
    y_max = a * x_min + b
    print(f"a = {a} < 0  →  støtten vendes om")
    print(f"y_min = {a} · {x_max} + ({b}) = {y_min}")
    print(f"y_max = {a} · {x_min} + ({b}) = {y_max}")
print(f"Støtteinterval for Y: [{y_min}, {y_max}]")

print()
print("═" * 55)
print("TRIN 3 — Tæthedsfunktionen f_Y(y)")
print("═" * 55)
y = symbols('y', real=True)

print(f"Tæthedsfunktionen for X:")
print(f"  f_X(x) = {f_X_sym}")
print()

x_af_y = (y - b) / a
print(f"Invers transformation: x = (y - b) / a = (y - {b}) / {a}")
print(f"  x(y) = {x_af_y} = {simplify(x_af_y)}")
print()

f_Y_raa = f_X_sym.subs(x, x_af_y) * Abs(Rational(1, a))
f_Y_sym = simplify(f_Y_raa)
print(f"Indsæt i formlen  f_Y(y) = f_X((y-b)/a) · 1/|a|:")
print(f"  f_Y(y) = f_X({simplify(x_af_y)}) · 1/|{a}|")
print(f"  f_Y(y) = {expand(f_Y_sym)}")
print()

print("Stykvis definition:")
print(f"  f_Y(y) = 0                        for y < {y_min}")
print(f"  f_Y(y) = {expand(f_Y_sym)}")
print(f"           for {y_min} ≤ y ≤ {y_max}")
print(f"  f_Y(y) = 0                        for y > {y_max}")

print()
print("═" * 55)
print("TRIN 4 — Verificering (∫ f_Y(y) dy = 1)")
print("═" * 55)
integral_Y = integrate(f_Y_sym, (y, y_min, y_max))
integral_Y_simplified = simplify(integral_Y)
print(f"∫ f_Y(y) dy fra {y_min} til {y_max} = {integral_Y_simplified}")
if abs(float(integral_Y_simplified) - 1.0) < 1e-6:
    print("✅ f_Y(y) er en gyldig tæthedsfunktion (integral = 1)")
else:
    print(f"❌ Integral = {float(integral_Y_simplified):.6f} — tjek din f_X(x)")
```

## 2.4 Middelværdi og varians — diskret

Bruges til at beregne E[X] og Var(X) for en diskret stokastisk variabel.

**Indsæt:**
- `x_vaerdier` = liste med mulige værdier af X
- `sandsynligheder` = tilhørende sandsynligheder (skal summe til 1)

$$E[X] = \sum x \cdot f_X(x), \quad \text{Var}(X) = E[X^2] - (E[X])^2$$

```python
# ── INDSÆT DINE TAL HER ──────────────────────────
x_vaerdier      = [-1,   1,   2,   4  ]   # mulige værdier
sandsynligheder = [1/6, 1/3, 1/6, 1/3]   # tilhørende sandsynligheder
# ─────────────────────────────────────────────────

x = np.array(x_vaerdier)
p = np.array(sandsynligheder)

print(f"Sum af sandsynligheder = {p.sum():.4f}", "✅" if abs(p.sum()-1)<1e-9 else "❌")
print()

E_X   = np.sum(x * p)
E_X2  = np.sum(x**2 * p)
Var_X = E_X2 - E_X**2

print(f"E[X]   = Σ x·P(x)   = {E_X:.4f}")
print(f"E[X²]  = Σ x²·P(x)  = {E_X2:.4f}")
print(f"Var(X) = {E_X2:.4f} - {E_X**2:.4f} = {Var_X:.4f}")
```

## 2.5 Marginalfordeling og uafhængighed — diskret simultant PMF

Bruges til at finde marginale PMF'er fra en simultant PMF-tabel, og teste om X og Y er uafhængige.

**Indsæt:**
- `joint` = matrix med simultane sandsynligheder (rækker = x-værdier, kolonner = y-værdier)
- `x_vals`, `y_vals` = de mulige værdier af X og Y

```python
# ── INDSÆT DINE TAL HER ──────────────────────────
# Rækker = x-værdier (-1, 0, 1), kolonner = y-værdier (1, 2, 3, 4)
joint = np.array([
    [0.1,  0.08, 0.02],   # x = 0
    [0.12, 0.26, 0.06],   # x = 1
    [0.05, 0.09, 0.04],   # x = 2  ← fjern 0.03
    [0.08, 0.07, 0.03],   # x = 3
])

x_vals = [0, 1, 2, 3]
y_vals = [0, 1, 2]
# ─────────────────────────────────────────────────

f_X = joint.sum(axis=1)   # summer over kolonner (y)
f_Y = joint.sum(axis=0)   # summer over rækker (x)

print(f"Sum af alle sandsynligheder = {joint.sum():.4f}", "✅" if abs(joint.sum()-1)<1e-9 else "❌")
print()
print("Marginal PMF for X:")
for xv, px in zip(x_vals, f_X):
    print(f"  f_X({xv:2}) = {px:.4f}")
print(f"  Sum = {f_X.sum():.4f}", "✅")
print()
print("Marginal PMF for Y:")
for yv, py in zip(y_vals, f_Y):
    print(f"  f_Y({yv}) = {py:.4f}")
print(f"  Sum = {f_Y.sum():.4f}", "✅")
print()

# Uafhængighedstest
uafh = True
for i, xv in enumerate(x_vals):
    for j, yv in enumerate(y_vals):
        produkt = f_X[i] * f_Y[j]
        faktisk = joint[i, j]
        if abs(produkt - faktisk) > 1e-9:
            print(f"❌ IKKE uafhængige: f({xv},{yv}) = {faktisk:.4f} ≠ f_X({xv})·f_Y({yv}) = {produkt:.4f}")
            uafh = False
            break
    if not uafh:
        break
if uafh:
    print("✅ X og Y er uafhængige")
```

## 2.6 Find konstant c i joint PDF

Bruges til at finde konstanten c der gør en joint PDF gyldig (dobbeltintegral = 1).

**Indsæt:**
- `f_uden_c` = joint PDF uden konstanten c (sæt c = 1 i formlen)
- `x_min`, `x_max`, `y_min`, `y_max` = integrationsgrænserne

$$\int\!\int c \cdot f(x,y)\,dy\,dx = 1 \implies c = \frac{1}{\int\!\int f(x,y)\,dy\,dx}$$

```python
from scipy.integrate import dblquad

# ── INDSÆT DINE TAL HER ──────────────────────────
def f_uden_c(y, x):          # NB: dblquad tager (y, x) som argument!
    return x * (1 + y)       # ← ændr til din funktion UDEN c

x_min, x_max = 0, 2          # grænser for x
y_min, y_max = 0, 1          # grænser for y
# ─────────────────────────────────────────────────

integral, _ = dblquad(f_uden_c, x_min, x_max,
                       lambda x: y_min, lambda x: y_max)
c = 1 / integral

print(f"∫∫ f(x,y) dy dx (uden c) = {integral:.6f}")
print(f"c = 1 / {integral:.6f} = {c:.6f}")
print()

# Verificer
def f_med_c(y, x):
    return c * x * (1 + y)    # ← ændr til c · din funktion

tjek, _ = dblquad(f_med_c, x_min, x_max,
                   lambda x: y_min, lambda x: y_max)
print(f"Verificering: ∫∫ c·f(x,y) dy dx = {tjek:.6f}", "✅" if abs(tjek-1)<1e-6 else "❌")
```

# 3. Statistik

## 3.1 Parret t-test

Bruges når **samme enhed** er målt to gange (fx samme bygning før/efter, samme person før/efter).
Vi danner differencer D_i og tester om middelværdien er nul.

**Indsæt:**
- `foer` = liste med målinger før
- `efter` = liste med målinger efter
- `alpha` = signifikansniveau (fx 0.05 for 5%)
- `ensidet` = True hvis H₁: δ > 0, False hvis H₁: δ ≠ 0

$$D_i = X_i - Y_i, \quad t = \frac{\bar{D} - \delta_0}{s_D / \sqrt{n}} \sim t_{n-1}$$

```python
# ── INDSÆT DINE TAL HER ──────────────────────────
foer  = np.array([158,162,171,155,168,160,175,153,164,170,159])
efter = np.array([142,149,158,145,152,148,161,144,150,155,147])
alpha   = 0.05    # signifikansniveau
ensidet = True    # True = ensidet (H₁: δ > 0), False = tosidet (H₁: δ ≠ 0)
delta_0 = 0       # nulhypotese-værdien (næsten altid 0)
# ─────────────────────────────────────────────────

D    = foer - efter
n    = len(D)
D_bar = np.mean(D)
s_D  = np.std(D, ddof=1)
df   = n - 1
SE   = s_D / np.sqrt(n)
t    = (D_bar - delta_0) / SE

# p-værdi
if ensidet:
    p_vardi = 1 - stats.t.cdf(t, df)  # ensidet: H₁: δ > 0
else:
    p_vardi = 2 * (1 - stats.t.cdf(abs(t), df))  # tosidet

# Kritisk t-værdi
if ensidet:
    t_kritisk = stats.t.ppf(1 - alpha, df)
else:
    t_kritisk = stats.t.ppf(1 - alpha/2, df)

print("═" * 45)
print("PARRET T-TEST")
print("═" * 45)
print(f"Differencer D_i = Før - Efter:")
for i, d in enumerate(D, 1):
    print(f"  D_{i:02} = {d:.2f}")
print()
print(f"n     = {n}")
print(f"D̄     = {D_bar:.4f}")
print(f"s_D   = {s_D:.4f}")
print(f"df    = {df}")
print(f"SE    = {SE:.4f}")
print(f"t     = {t:.4f}")
print(f"t*    = {t_kritisk:.4f}  ({'ensidet' if ensidet else 'tosidet'}, α={alpha})")
print(f"p-værdi = {p_vardi:.6f}")
print()
if abs(t) > t_kritisk:
    print(f"✅ |t| = {abs(t):.3f} > t* = {t_kritisk:.3f}  →  FORKAST H₀")
else:
    print(f"❌ |t| = {abs(t):.3f} ≤ t* = {t_kritisk:.3f}  →  BEHOLD H₀")
```

## 3.2 Konfidensinterval — parret test

KI beregnes altid med den **tosidede** kritiske t-værdi, selv når testen er ensidet.

**Indsæt:**
- `D_bar` = middelværdi af differencer
- `s_D` = standardafvigelse af differencer
- `n` = antal par
- `konfidens` = fx 0.95 for 95% KI

$$KI = \bar{D} \pm t^* \cdot \frac{s_D}{\sqrt{n}}$$

```python
# ── INDSÆT DINE TAL HER ──────────────────────────
D_bar     = 13.0909    # middelværdi af differencer
s_D       = 2.2563   # standardafvigelse af differencer
n         = 11      # antal par
konfidens = 0.95    # fx 0.95 for 95% KI
# ─────────────────────────────────────────────────

df        = n - 1
alpha     = 1 - konfidens
t_kritisk = stats.t.ppf(1 - alpha/2, df)   # altid tosidet for KI
margin    = t_kritisk * s_D / np.sqrt(n)

KI_nedre = D_bar - margin
KI_ovre  = D_bar + margin

print(f"df        = {df}")
print(f"t*        = {t_kritisk:.4f}  (tosidet, α={alpha:.2f})")
print(f"Margin    = {t_kritisk:.4f} × {s_D}/{np.sqrt(n):.4f} = {margin:.4f}")
print()
print(f"{int(konfidens*100)}% Konfidensinterval:")
print(f"  Nedre grænse = {D_bar:.4f} - {margin:.4f} = {KI_nedre:.4f}")
print(f"  Øvre grænse  = {D_bar:.4f} + {margin:.4f} = {KI_ovre:.4f}")
print(f"  KI = ({KI_nedre:.3f} ; {KI_ovre:.3f})")
print()
if KI_nedre > 0:
    print("✅ 0 er IKKE i intervallet — understøtter forkastelse af H₀")
else:
    print("❌ 0 er i intervallet — understøtter IKKE forkastelse af H₀")
```

## 3.3 Uparret t-test (pooled varians)

Bruges når **to uafhængige grupper** sammenlignes og variansen antages ens i begge grupper.

**Indsæt:**
- `gruppe1`, `gruppe2` = de to datasæt som lister
- `alpha` = signifikansniveau
- `ensidet` = True hvis H₁: μ₂ − μ₁ > 0, False hvis H₁: μ₂ − μ₁ ≠ 0

$$s_p^2 = \frac{(n_1-1)s_1^2 + (n_2-1)s_2^2}{n_1+n_2-2}, \quad SE = s_p\sqrt{\frac{1}{n_1}+\frac{1}{n_2}}, \quad t = \frac{\bar{x}_2 - \bar{x}_1}{SE}$$

```python
# ── INDSÆT DINE TAL HER ──────────────────────────
gruppe1 = np.array([312, 298, 325, 308, 319, 302, 315, 296])  # gruppe 1
gruppe2 = np.array([334, 318, 341, 329, 322, 338, 327, 315])  # gruppe 2
alpha   = 0.05    # signifikansniveau
ensidet = True    # True = ensidet (H₁: μ₂ > μ₁), False = tosidet
# ─────────────────────────────────────────────────

n1, n2   = len(gruppe1), len(gruppe2)
x1_bar   = np.mean(gruppe1)
x2_bar   = np.mean(gruppe2)
s1       = np.std(gruppe1, ddof=1)
s2       = np.std(gruppe2, ddof=1)

sp2  = ((n1-1)*s1**2 + (n2-1)*s2**2) / (n1+n2-2)
sp   = np.sqrt(sp2)
SE   = sp * np.sqrt(1/n1 + 1/n2)
df   = n1 + n2 - 2
t    = (x2_bar - x1_bar) / SE

if ensidet:
    p_vardi   = 1 - stats.t.cdf(t, df)
    t_kritisk = stats.t.ppf(1 - alpha, df)
else:
    p_vardi   = 2 * (1 - stats.t.cdf(abs(t), df))
    t_kritisk = stats.t.ppf(1 - alpha/2, df)

print("═" * 45)
print("UPARRET T-TEST (POOLED VARIANS)")
print("═" * 45)
print(f"n₁ = {n1},  x̄₁ = {x1_bar:.4f},  s₁ = {s1:.4f}")
print(f"n₂ = {n2},  x̄₂ = {x2_bar:.4f},  s₂ = {s2:.4f}")
print()
print(f"s²_p  = {sp2:.4f}")
print(f"s_p   = {sp:.4f}")
print(f"SE    = {SE:.4f}")
print(f"df    = {df}")
print(f"t     = {t:.4f}")
print(f"t*    = {t_kritisk:.4f}  ({'ensidet' if ensidet else 'tosidet'}, α={alpha})")
print(f"p-værdi = {p_vardi:.6f}")
print()
if abs(t) > t_kritisk:
    print(f"✅ |t| = {abs(t):.3f} > t* = {t_kritisk:.3f}  →  FORKAST H₀")
else:
    print(f"❌ |t| = {abs(t):.3f} ≤ t* = {t_kritisk:.3f}  →  BEHOLD H₀")
```

## 3.4 Konfidensinterval — uparret test

**Indsæt:**
- `x1_bar`, `x2_bar` = middelværdier for de to grupper
- `sp` = pooled standardafvigelse
- `n1`, `n2` = stikprøvestørrelser
- `konfidens` = fx 0.95 for 95% KI

$$KI = (\bar{x}_2 - \bar{x}_1) \pm t^* \cdot SE$$

```python
# ── INDSÆT DINE TAL HER ──────────────────────────
x1_bar    = 309.375   # middelværdi gruppe 1
x2_bar    = 328.0     # middelværdi gruppe 2
sp        = 9.827     # pooled standardafvigelse
n1        = 8         # n for gruppe 1
n2        = 8         # n for gruppe 2
konfidens = 0.95      # fx 0.95 for 95% KI
# ─────────────────────────────────────────────────

df        = n1 + n2 - 2
alpha     = 1 - konfidens
SE        = sp * np.sqrt(1/n1 + 1/n2)
t_kritisk = stats.t.ppf(1 - alpha/2, df)
forskel   = x2_bar - x1_bar
margin    = t_kritisk * SE

KI_nedre = forskel - margin
KI_ovre  = forskel + margin

print(f"x̄₂ - x̄₁  = {forskel:.4f}")
print(f"SE        = {SE:.4f}")
print(f"df        = {df}")
print(f"t*        = {t_kritisk:.4f}  (tosidet, α={alpha:.2f})")
print(f"Margin    = {margin:.4f}")
print()
print(f"{int(konfidens*100)}% Konfidensinterval:")
print(f"  Nedre grænse = {forskel:.4f} - {margin:.4f} = {KI_nedre:.4f}")
print(f"  Øvre grænse  = {forskel:.4f} + {margin:.4f} = {KI_ovre:.4f}")
print(f"  KI = ({KI_nedre:.3f} ; {KI_ovre:.3f})")
print()
if KI_nedre > 0:
    print("✅ 0 er IKKE i intervallet — understøtter forkastelse af H₀")
else:
    print("❌ 0 er i intervallet — understøtter IKKE forkastelse af H₀")
```

## 3.5 Lineær regression (OLS)

Bruges til at fitte en ret linje y = β₀ + β₁x til data og teste om der er en signifikant lineær sammenhæng.

**Indsæt:**
- `x` = forklarende variabel (fx vægt, forespørgsler)
- `y` = afhængig variabel (fx spænding, responstid)
- `alpha` = signifikansniveau for hypotesetest

$$\hat{\beta}_1 = \frac{\sum(x_i-\bar{x})(y_i-\bar{y})}{\sum(x_i-\bar{x})^2}, \quad \hat{\beta}_0 = \bar{y} - \hat{\beta}_1\bar{x}$$
$$R^2 = 1 - \frac{SS_{res}}{SS_{tot}}$$

```python
# ── INDSÆT DINE TAL HER ──────────────────────────
x = np.array([18, 20, 22, 24, 26, 28, 30, 32])
y = np.array([3, 5, 6, 8, 10, 11, 14, 15])
alpha = 0.05   # signifikansniveau
# ─────────────────────────────────────────────────

n      = len(x)
x_bar  = np.mean(x)
y_bar  = np.mean(y)
S_xx   = np.sum((x - x_bar)**2)
S_xy   = np.sum((x - x_bar)*(y - y_bar))

beta1  = S_xy / S_xx
beta0  = y_bar - beta1 * x_bar
y_hat  = beta0 + beta1 * x

SS_res = np.sum((y - y_hat)**2)
SS_tot = np.sum((y - y_bar)**2)
R2     = 1 - SS_res / SS_tot

s2     = SS_res / (n - 2)
SE_b1  = np.sqrt(s2 / S_xx)
t_b1   = beta1 / SE_b1
p_b1   = 2 * (1 - stats.t.cdf(abs(t_b1), df=n-2))
t_krit = stats.t.ppf(1 - alpha/2, df=n-2)

print("═" * 45)
print("LINEÆR REGRESSION")
print("═" * 45)
print(f"β̂₀ (skæring)  = {beta0:.4f}")
print(f"β̂₁ (hældning) = {beta1:.4f}")
print(f"Model: ŷ = {beta0:.4f} + {beta1:.4f}·x")
print()
print(f"R²            = {R2:.4f}  →  {R2*100:.2f}% af variationen forklares")
print()
print(f"--- Hypotesetest for β₁ (H₀: β₁=0) ---")
print(f"SE(β̂₁)  = {SE_b1:.4f}")
print(f"t       = {t_b1:.4f}")
print(f"t*      = {t_krit:.4f}  (tosidet, α={alpha}, df={n-2})")
print(f"p-værdi = {p_b1:.6f}")
print()
if abs(t_b1) > t_krit:
    print(f"✅ Signifikant lineær sammenhæng (p={p_b1:.4f} < {alpha})")
else:
    print(f"❌ Ingen signifikant lineær sammenhæng (p={p_b1:.4f} ≥ {alpha})")
```

## 3.5b Hypotesetest for β₁ — trin-for-trin

Bruges til at beregne t-statistik og p-værdi for H₀: β₁ = 0 ud fra allerede kendte regressionsstørrelser, uden at genberegne hele modellen.

**Indsæt:**
- `beta1` = den estimerede hældning β̂₁
- `SS_res` = residualsum af kvadrater
- `S_xx` = Σ(xᵢ − x̄)²
- `n` = antal observationer
- `alpha` = signifikansniveau

$$S_{xx} = \sum(x_i - \bar{x})^2, \quad s = \sqrt{\frac{SS_{res}}{n-2}}, \quad SE(\hat{\beta}_1) = \frac{s}{\sqrt{S_{xx}}}$$

$$t = \frac{\hat{\beta}_1}{SE(\hat{\beta}_1)}, \quad p = 2\bigl(1 - t_{cdf}(|t|,\; n-2)\bigr)$$

```python
# ── INDSÆT DINE TAL HER ──────────────────────────
beta1  = 0.8690    # estimeret hældning β̂₁
SS_res = 0.8452    # residualsum af kvadrater Σ(yᵢ - ŷᵢ)²
S_xx   = 168.0     # Σ(xᵢ - x̄)²
n      = 8         # antal observationer
alpha  = 0.05      # signifikansniveau
# ─────────────────────────────────────────────────

from scipy import stats
import numpy as np

df = n - 2

print("═" * 50)
print("TRIN 1 — Residualstandardafvigelse s")
print("═" * 50)
s = np.sqrt(SS_res / df)
print(f"s = √(SS_res / (n-2))")
print(f"s = √({SS_res} / {df})")
print(f"s = √{SS_res/df:.6f}")
print(f"s = {s:.4f}")

print()
print("═" * 50)
print("TRIN 2 — Standardfejl SE(β̂₁)")
print("═" * 50)
SE_b1 = s / np.sqrt(S_xx)
print(f"SE(β̂₁) = s / √S_xx")
print(f"SE(β̂₁) = {s:.4f} / √{S_xx}")
print(f"SE(β̂₁) = {s:.4f} / {np.sqrt(S_xx):.4f}")
print(f"SE(β̂₁) = {SE_b1:.4f}")

print()
print("═" * 50)
print("TRIN 3 — t-statistik")
print("═" * 50)
t = beta1 / SE_b1
print(f"t = β̂₁ / SE(β̂₁)")
print(f"t = {beta1} / {SE_b1:.4f}")
print(f"t = {t:.4f}")

print()
print("═" * 50)
print("TRIN 4 — p-værdi og konklusion")
print("═" * 50)
p = 2 * (1 - stats.t.cdf(abs(t), df))
t_krit = stats.t.ppf(1 - alpha/2, df)
print(f"df    = n - 2 = {df}")
print(f"p     = 2 · (1 - t_cdf(|{t:.4f}|, {df}))")
print(f"p     = {p:.6f}")
print(f"t*    = {t_krit:.4f}  (tosidet, α={alpha}, df={df})")
print()
if abs(t) > t_krit:
    print(f"✅ |t| = {abs(t):.4f} > t* = {t_krit:.4f}  →  FORKAST H₀")
    print(f"   β₁ er signifikant forskellig fra 0 (p = {p:.4f} < {alpha})")
else:
    print(f"❌ |t| = {abs(t):.4f} ≤ t* = {t_krit:.4f}  →  BEHOLD H₀")
    print(f"   Ingen signifikant lineær sammenhæng (p = {p:.4f} ≥ {alpha})")
```

## 3.6 Én-stikprøve t-test

Bruges til at teste om en stikprøves middelværdi stemmer overens med en specificeret værdi μ₀.

**Indsæt:**
- `data` = stikprøven som liste
- `mu_0` = den specificerede middelværdi vi tester imod
- `alpha` = signifikansniveau
- `ensidet` = True eller False

$$t = \frac{\bar{x} - \mu_0}{s/\sqrt{n}} \sim t_{n-1} \text{ under } H_0$$

```python
# ── INDSÆT DINE TAL HER ──────────────────────────
data    = np.array([3148, 2956, 2803, 2933, 2869, 3111, 2789, 2995, 2909, 2929, 3148, 2867])
mu_0    = 3000    # den specificerede middelværdi (H₀: μ = mu_0)
alpha   = 0.05    # signifikansniveau
ensidet = False   # True = ensidet, False = tosidet
# ─────────────────────────────────────────────────

n      = len(data)
x_bar  = np.mean(data)
s      = np.std(data, ddof=1)
df     = n - 1
SE     = s / np.sqrt(n)
t      = (x_bar - mu_0) / SE

if ensidet:
    p_vardi   = 1 - stats.t.cdf(t, df)
    t_kritisk = stats.t.ppf(1 - alpha, df)
else:
    p_vardi   = 2 * (1 - stats.t.cdf(abs(t), df))
    t_kritisk = stats.t.ppf(1 - alpha/2, df)

print("═" * 45)
print("ÉN-STIKPRØVE T-TEST")
print("═" * 45)
print(f"n      = {n}")
print(f"x̄      = {x_bar:.4f}")
print(f"μ₀     = {mu_0}")
print(f"s      = {s:.4f}")
print(f"SE     = {SE:.4f}")
print(f"df     = {df}")
print(f"t      = {t:.4f}")
print(f"t*     = {t_kritisk:.4f}  ({'ensidet' if ensidet else 'tosidet'}, α={alpha})")
print(f"p-værdi = {p_vardi:.6f}")
print()
if abs(t) > t_kritisk:
    print(f"✅ FORKAST H₀: middelværdi er signifikant forskellig fra {mu_0}")
else:
    print(f"❌ BEHOLD H₀: ingen signifikant forskel fra {mu_0}")
```

```python
# ══════════════════════════════════════════════════
# INDSÆT DINE TAL HER
# ══════════════════════════════════════════════════
gruppe1   = np.array([42.1, 38.7, 51.3, 44.8, 39.2, 47.6, 43.3, 50.1, 41.5, 45.9])
gruppe2   = np.array([38.4, 35.9, 46.2, 41.7, 37.1, 43.8, 40.2, 46.9, 38.8, 42.4])
test_type = "parret"   # "parret" eller "uparret"
ensidet   = True       # True = ensidet, False = tosidet
konfidens = 0.95       # fx 0.95 for 95% KI
alpha     = 1 - konfidens
# ══════════════════════════════════════════════════

print("═" * 50)
print(f"TEST: {test_type.upper()} — {'ENSIDET' if ensidet else 'TOSIDET'}")
print(f"Signifikansniveau: α = {alpha},  {int(konfidens*100)}% KI")
print("═" * 50)

if test_type == "parret":
    D     = gruppe1 - gruppe2
    n     = len(D)
    D_bar = np.mean(D)
    s_D   = np.std(D, ddof=1)
    df    = n - 1
    SE    = s_D / np.sqrt(n)
    t     = D_bar / SE
    print(f"n = {n},  D̄ = {D_bar:.4f},  s_D = {s_D:.4f}")
else:
    n1, n2 = len(gruppe1), len(gruppe2)
    x1_bar = np.mean(gruppe1)
    x2_bar = np.mean(gruppe2)
    s1     = np.std(gruppe1, ddof=1)
    s2     = np.std(gruppe2, ddof=1)
    sp2    = ((n1-1)*s1**2 + (n2-1)*s2**2) / (n1+n2-2)
    sp     = np.sqrt(sp2)
    SE     = sp * np.sqrt(1/n1 + 1/n2)
    df     = n1 + n2 - 2
    t      = (x2_bar - x1_bar) / SE
    D_bar  = x2_bar - x1_bar
    print(f"n₁={n1}, x̄₁={x1_bar:.3f}, s₁={s1:.3f}")
    print(f"n₂={n2}, x̄₂={x2_bar:.3f}, s₂={s2:.3f}")
    print(f"s_p = {sp:.4f},  SE = {SE:.4f}")

# t-test
if ensidet:
    p_vardi   = 1 - stats.t.cdf(t, df)
    t_kritisk = stats.t.ppf(1 - alpha, df)
else:
    p_vardi   = 2 * (1 - stats.t.cdf(abs(t), df))
    t_kritisk = stats.t.ppf(1 - alpha/2, df)

print(f"df = {df},  t = {t:.4f},  t* = {t_kritisk:.4f},  p = {p_vardi:.6f}")
print()
if abs(t) > t_kritisk:
    print(f"✅ FORKAST H₀  (|t|={abs(t):.3f} > t*={t_kritisk:.3f})")
else:
    print(f"❌ BEHOLD H₀   (|t|={abs(t):.3f} ≤ t*={t_kritisk:.3f})")

# KI
t_ki     = stats.t.ppf(1 - alpha/2, df)
margin   = t_ki * SE
KI_nedre = D_bar - margin
KI_ovre  = D_bar + margin
print()
print(f"{int(konfidens*100)}% KI = ({KI_nedre:.4f} ; {KI_ovre:.4f})")
if KI_nedre > 0:
    print("✅ 0 er IKKE i intervallet")
else:
    print("❌ 0 er i intervallet")
```
